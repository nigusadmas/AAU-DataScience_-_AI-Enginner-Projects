```python
import warnings
warnings.filterwarnings("ignore")

from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.neighbors import NearestNeighbors
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import precision_score, recall_score

# ------------------------------------------------------------
# 0. LOAD DATA
# ------------------------------------------------------------

DATA_DIR = Path(".")  # Change this if your CSV files are elsewhere.

daily = pd.read_csv(DATA_DIR / "daily_sales.csv")
reference = pd.read_csv(DATA_DIR / "product_reference.csv")

daily["date"] = pd.to_datetime(daily["date"])
daily = daily.sort_values(["product_id", "date"]).reset_index(drop=True)

print("Daily sales shape:", daily.shape)
print("Reference shape:", reference.shape)
print("\nProduct reference:")
print(reference.to_string(index=False))

print("\nBasic data quality:")
print("Missing values:\n", daily.isna().sum())
print("Duplicate product/date rows:", daily.duplicated(["product_id", "date"]).sum())

# ------------------------------------------------------------
# 0B. INITIAL PRODUCT PROFILE
# ------------------------------------------------------------

profile = daily.groupby("product_id")["sales"].agg(
    count="count",
    mean="mean",
    std="std",
    median="median",
    min="min",
    max="max"
)
print("\nProduct profile:")
print(profile.round(3))
```

## EXERCISE 1 — Z-SCORE DETECTION

```python
# EXERCISE 1 — Z-SCORE DETECTION
```

```python
# ------------------------------------------------------------
# 4–5. Compute mean, std, Z-score independently per product
# ------------------------------------------------------------

z_df = daily.copy()

product_stats = z_df.groupby("product_id")["sales"].agg(
    mean_sales="mean",
    std_sales="std"
)

z_df = z_df.join(product_stats, on="product_id")

# Z = (x - mean) / standard deviation
z_df["z_score"] = (
    (z_df["sales"] - z_df["mean_sales"]) / z_df["std_sales"]
)

# Anomaly candidate if absolute Z-score > 3
z_df["z_flag"] = z_df["z_score"].abs() > 3

print("\nExercise 1 — Z-score counts")
z_counts = z_df.groupby("product_id")["z_flag"].sum().astype(int)
print(z_counts)

print("\nZ-score flagged rows:")
print(
    z_df[z_df["z_flag"]]
    [["product_id", "date", "sales", "z_score"]]
    .sort_values(["product_id", "date"])
    .to_string(index=False)
)

# ------------------------------------------------------------
# 6. P002 — GLOBAL VS LOCAL BASELINE
# ------------------------------------------------------------
# IMPORTANT:
# The exercise refers to the injected level-shift date.
# If you have anomaly_answer_key.csv, use its level_shift_start date.
# Otherwise, the code below finds a candidate change point using a
# simple two-segment mean comparison. This is a diagnostic aid,
# not a replacement for the ground-truth key.

def find_candidate_shift_date(group, min_segment_days=90):
    g = group.sort_values("date").copy()

    # Remove the obvious extreme value so it does not dominate the search.
    median = g["sales"].median()
    mad = np.median(np.abs(g["sales"] - median))
    if mad > 0:
        clean = g[(g["sales"] - median).abs() <= 50 * mad].copy()
    else:
        clean = g.copy()

    x = clean["sales"].to_numpy(dtype=float)
    dates = clean["date"].to_numpy()

    n = len(x)
    if n < 2 * min_segment_days:
        return None

    csum = np.cumsum(x)
    csum2 = np.cumsum(x ** 2)

    best = None

    for i in range(min_segment_days, n - min_segment_days):
        n1, n2 = i, n - i

        mean1 = csum[i - 1] / n1
        mean2 = (csum[-1] - csum[i - 1]) / n2

        sse1 = csum2[i - 1] - n1 * mean1**2
        sse2 = (csum2[-1] - csum2[i - 1]) - n2 * mean2**2

        score = sse1 + sse2

        if best is None or score < best[0]:
            best = (score, i, mean1, mean2)

    return pd.Timestamp(dates[best[1]]), best[2], best[3]


p002 = daily[daily["product_id"] == "P002"].copy()

candidate = find_candidate_shift_date(p002)

if candidate is not None:
    candidate_date, before_mean_candidate, after_mean_candidate = candidate
    print("\nCandidate P002 level-shift date:", candidate_date.date())
    print("Candidate before mean:", round(before_mean_candidate, 2))
    print("Candidate after mean :", round(after_mean_candidate, 2))
else:
    candidate_date = None
    print("\nNo reliable candidate shift date found.")

# If your answer key gives the exact injected shift date,
# replace candidate_date with that date, e.g.:
# candidate_date = pd.Timestamp("YYYY-MM-DD")

if candidate_date is not None:
    p002_before = p002[p002["date"] < candidate_date]["sales"]
    p002_after = p002[p002["date"] >= candidate_date]["sales"]

    global_mean = p002["sales"].mean()
    global_std = p002["sales"].std()

    print("\nP002 baseline comparison")
    print("GLOBAL:")
    print(" mean =", round(global_mean, 3))
    print(" std  =", round(global_std, 3))

    print("\nBEFORE shift:")
    print(" mean =", round(p002_before.mean(), 3))
    print(" std  =", round(p002_before.std(), 3))

    print("\nAFTER shift:")
    print(" mean =", round(p002_after.mean(), 3))
    print(" std  =", round(p002_after.std(), 3))

    # Show how the same post-shift values look under the global baseline.
    print("\nInterpretation:")
    print(
        "Once the post-shift period becomes part of the global baseline, "
        "the global mean/std move toward the new level. Therefore ordinary "
        "post-shift values can receive small Z-scores and stop looking anomalous."
    )


# ------------------------------------------------------------
# 7. P003 — DISTRIBUTIONAL SHAPE
# ------------------------------------------------------------

p003 = daily.loc[daily["product_id"] == "P003", "sales"]

print("\nP003 distribution summary:")
print(p003.describe())

plt.figure(figsize=(8, 5))
plt.hist(p003, bins=20, edgecolor="black")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.title("P003 Sales Distribution")
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()

print(
    "\nP003 interpretation: the distribution is highly concentrated at zero "
    "and has a long right tail. It is strongly skewed rather than normal/symmetric."
)

# ------------------------------------------------------------
# 8. THRESHOLD SENSITIVITY — |Z| > 2, 3, 4
# ------------------------------------------------------------

threshold_results = []

for threshold in [2, 3, 4]:
    temp = z_df.assign(flag=z_df["z_score"].abs() > threshold)
    counts = temp.groupby("product_id")["flag"].sum()

    for product, count in counts.items():
        threshold_results.append(
            {
                "threshold": threshold,
                "product_id": product,
                "flagged_count": int(count)
            }
        )

threshold_table = pd.DataFrame(threshold_results)

print("\nExercise 1 threshold sensitivity:")
print(
    threshold_table
    .pivot(index="product_id", columns="threshold", values="flagged_count")
    .rename(columns={2: "Z>2", 3: "Z>3", 4: "Z>4"})
)
```

## EXERCISE 2 — MODIFIED Z-SCORE (MAD)

```python
# EXERCISE 2 — MODIFIED Z-SCORE (MAD)
```

```python
# ------------------------------------------------------------
# 9–10. Median, MAD and Modified Z-score
# ------------------------------------------------------------

mz_df = daily.copy()

median_by_product = mz_df.groupby("product_id")["sales"].median()
mz_df["median_sales"] = mz_df["product_id"].map(median_by_product)

mz_df["absolute_deviation"] = (
    mz_df["sales"] - mz_df["median_sales"]
).abs()

mad_by_product = (
    mz_df.groupby("product_id")["absolute_deviation"]
    .median()
    .rename("MAD")
)

mz_df = mz_df.join(mad_by_product, on="product_id")

# Formula:
# Modified Z = 0.6745 * (x - median) / MAD
#
# A MAD of zero is possible for highly discrete/intermittent data.
# P003 is an important example. In that case the ordinary formula
# is undefined and must NOT be silently replaced with a fake number.

mz_df["modified_z"] = np.nan

valid_mad = mz_df["MAD"] > 0

mz_df.loc[valid_mad, "modified_z"] = (
    0.6745
    * (mz_df.loc[valid_mad, "sales"] - mz_df.loc[valid_mad, "median_sales"])
    / mz_df.loc[valid_mad, "MAD"]
)

mz_df["mz_flag"] = mz_df["modified_z"].abs() > 3.5

print("\nExercise 2 — Median and MAD:")
print(
    mz_df.groupby("product_id")
    .agg(
        median=("median_sales", "first"),
        MAD=("MAD", "first"),
        flagged=("mz_flag", "sum")
    )
    .round(3)
)

print("\nImportant P003 note:")
print(
    "If P003 has MAD = 0, the standard modified-Z formula is undefined "
    "because division by zero occurs. This is a limitation of MAD-based "
    "detection for a distribution with a dominant exact median."
)

print("\nModified Z-score flagged rows where MAD > 0:")
print(
    mz_df[mz_df["mz_flag"]]
    [["product_id", "date", "sales", "median_sales", "MAD", "modified_z"]]
    .sort_values(["product_id", "date"])
    .to_string(index=False)
)


# ------------------------------------------------------------
# 11. ROBUSTNESS TEST
# ------------------------------------------------------------

def robust_statistics_for_extreme(product_id):
    s = daily.loc[daily["product_id"] == product_id, "sales"].astype(float)

    # "Most extreme" here means the largest absolute value.
    # For these data this identifies the obvious injected extreme.
    extreme_idx = s.abs().idxmax()
    extreme_value = s.loc[extreme_idx]

    with_point = s
    without_point = s.drop(extreme_idx)

    rows = []

    for label, series in [
        ("With extreme point", with_point),
        ("Without extreme point", without_point)
    ]:
        median = series.median()
        mad = np.median(np.abs(series - median))

        rows.append(
            {
                "condition": label,
                "mean": series.mean(),
                "std": series.std(),
                "median": median,
                "MAD": mad
            }
        )

    result = pd.DataFrame(rows)
    return extreme_idx, extreme_value, result


for product in ["P001", "P002"]:
    idx, extreme, robust_table = robust_statistics_for_extreme(product)

    print(f"\n{product} robustness test")
    print("Extreme row index:", idx)
    print("Extreme value:", extreme)
    print(robust_table.round(4).to_string(index=False))


# ------------------------------------------------------------
# 12. Z-SCORE VS MODIFIED Z-SCORE
# ------------------------------------------------------------

comparison = (
    z_df.groupby("product_id")["z_flag"].sum().rename("Z_score_flags")
    .to_frame()
    .join(
        mz_df.groupby("product_id")["mz_flag"]
        .sum()
        .rename("Modified_Z_flags")
    )
)

print("\nExercise 2 comparison:")
print(comparison.astype(int))

# Select up to 3 disagreement examples from products with valid MAD.
disagreement = mz_df.merge(
    z_df[["product_id", "date", "z_score", "z_flag"]],
    on=["product_id", "date"],
    how="left"
)

disagreement = disagreement[
    disagreement["z_flag"] != disagreement["mz_flag"]
]

print("\nExample rows where the methods disagree:")
print(
    disagreement[
        ["product_id", "date", "sales", "z_score", "modified_z", "z_flag", "mz_flag"]
    ]
    .sort_values(["product_id", "date"])
    .head(10)
    .to_string(index=False)
)

print(
    "\nInterpretation: an extreme point can inflate the mean and standard deviation, "
    "making its ordinary Z-score surprisingly small. Median and MAD resist that "
    "inflation, so Modified Z can flag the same point strongly."
)


# ------------------------------------------------------------
# 13. REFLECTION
# ------------------------------------------------------------

print(
    "\nExercise 2 reflection:\n"
    "A sustained level shift or gradual drift is difficult for both Z-score and "
    "Modified Z-score because both compare observations with one fixed global center. "
    "After a new level becomes common, observations from that new regime may no longer "
    "look unusual even though the transition itself is operationally important."
)
```

## EXERCISE 3 — KNN DISTANCE-BASED DETECTION

```python
# EXERCISE 3 — KNN DISTANCE-BASED DETECTION
```

```python
# ------------------------------------------------------------
# 14–16. PART A — UNIVARIATE KNN
# ------------------------------------------------------------

def univariate_knn_detector(df, k=5, top_fraction=0.03):
    results = []

    for product_id, g in df.groupby("product_id"):
        g = g.sort_values("date").copy()

        X = g[["sales"]].to_numpy()

        # n_neighbors=k+1 because each point is its own nearest neighbor
        # at distance 0. We remove that self-distance below.
        model = NearestNeighbors(n_neighbors=k + 1)
        model.fit(X)

        distances, _ = model.kneighbors(X)

        # Ignore column 0: it is the point itself.
        avg_distance = distances[:, 1:].mean(axis=1)

        g["knn_avg_distance"] = avg_distance

        n_flags = max(1, int(np.ceil(len(g) * top_fraction)))

        # Select exactly the top n rows instead of using >= threshold,
        # because ties can otherwise create far more than 3% flags.
        top_indices = np.argsort(-avg_distance, kind="mergesort")[:n_flags]

        g["knn_flag"] = False
        g.iloc[
            top_indices,
            g.columns.get_loc("knn_flag")
        ] = True

        results.append(g)

    return pd.concat(results, ignore_index=True)


knn_uni = univariate_knn_detector(
    daily,
    k=5,
    top_fraction=0.03
)

print("\nUnivariate KNN flagged counts:")
print(knn_uni.groupby("product_id")["knn_flag"].sum().astype(int))

print("\nTop univariate KNN anomalies:")
print(
    knn_uni[knn_uni["knn_flag"]]
    [["product_id", "date", "sales", "knn_avg_distance"]]
    .sort_values(["product_id", "knn_avg_distance"], ascending=[True, False])
    .to_string(index=False)
)

print(
    "\nComparison idea:\n"
    "On sales-only data, KNN is also driven by unusual numerical distances, "
    "so it often resembles a robust outlier detector. Differences occur because "
    "KNN uses local neighbor spacing rather than distance from one global center."
)


# ------------------------------------------------------------
# 17. PART B — MULTIVARIATE FEATURE ENGINEERING
# ------------------------------------------------------------

def build_time_features(df):
    feature_frames = []

    for product_id, g in df.groupby("product_id"):
        g = g.sort_values("date")

        # Create a true calendar grid so lag_1 means the previous calendar day
        # and lag_7 means the same weekday one week earlier.
        #
        # Duplicate product/date rows are aggregated only for feature construction.
        # We keep the original duplicate rows in the final scoring table.
        daily_series = (
            g.groupby("date")["sales"]
            .mean()
            .sort_index()
        )

        calendar = pd.date_range(
            daily_series.index.min(),
            daily_series.index.max(),
            freq="D"
        )

        s = daily_series.reindex(calendar)

        f = pd.DataFrame(index=calendar)

        f["sales_lag_1"] = s.shift(1)
        f["sales_lag_7"] = s.shift(7)

        # shift(1) is essential:
        # today's sales must NOT be included in today's trailing mean.
        f["rolling_mean_7"] = (
            s.shift(1)
            .rolling(7, min_periods=7)
            .mean()
        )

        f["day_of_week"] = calendar.dayofweek
        f["product_id"] = product_id
        f = f.reset_index().rename(columns={"index": "date"})

        feature_frames.append(f)

    feature_table = pd.concat(feature_frames, ignore_index=True)

    result = df.merge(
        feature_table,
        on=["product_id", "date"],
        how="left"
    )

    return result


multi_df = build_time_features(daily)

feature_cols = [
    "sales",
    "sales_lag_1",
    "sales_lag_7",
    "rolling_mean_7",
    "day_of_week"
]

print("\nEngineered features:")
print(multi_df[["product_id", "date"] + feature_cols].head(15).to_string(index=False))

print(
    "\nScaling explanation:\n"
    "KNN uses distance. If features have very different numerical scales, "
    "a large-scale feature can dominate the distance and hide useful information "
    "from smaller-scale features. StandardScaler puts the features on comparable scales."
)


# ------------------------------------------------------------
# 18–20. MULTIVARIATE KNN
# ------------------------------------------------------------

def multivariate_knn_detector(
    df,
    features,
    k=5,
    top_fraction=0.03
):
    results = []

    for product_id, g in df.groupby("product_id"):
        g = g.copy()

        valid = g[features].notna().all(axis=1)
        gv = g.loc[valid].copy()

        scaler = StandardScaler()
        X_scaled = scaler.fit_transform(gv[features])

        # k+1 because the point itself appears as the first neighbor.
        model = NearestNeighbors(n_neighbors=k + 1)
        model.fit(X_scaled)

        distances, _ = model.kneighbors(X_scaled)

        avg_distance = distances[:, 1:].mean(axis=1)

        gv["knn_avg_distance"] = avg_distance

        n_flags = max(1, int(np.ceil(len(gv) * top_fraction)))
        top_indices = np.argsort(-avg_distance, kind="mergesort")[:n_flags]

        gv["knn_flag"] = False
        gv.iloc[
            top_indices,
            gv.columns.get_loc("knn_flag")
        ] = True

        g["knn_avg_distance"] = np.nan
        g["knn_flag"] = False

        g.loc[gv.index, "knn_avg_distance"] = gv["knn_avg_distance"]
        g.loc[gv.index, "knn_flag"] = gv["knn_flag"]

        results.append(g)

    return pd.concat(results, ignore_index=True)


knn_multi = multivariate_knn_detector(
    multi_df,
    feature_cols,
    k=5,
    top_fraction=0.03
)

print("\nMultivariate KNN flagged counts:")
print(knn_multi.groupby("product_id")["knn_flag"].sum().astype(int))

print("\nTop multivariate KNN anomalies:")
print(
    knn_multi[knn_multi["knn_flag"]]
    [
        [
            "product_id",
            "date",
            "sales",
            "sales_lag_1",
            "sales_lag_7",
            "rolling_mean_7",
            "day_of_week",
            "knn_avg_distance"
        ]
    ]
    .sort_values(["product_id", "knn_avg_distance"], ascending=[True, False])
    .to_string(index=False)
)

print(
    "\nKey KNN idea:\n"
    "Z-score and Modified Z-score see only the raw sales value relative to one "
    "global center. Multivariate KNN can also see recent history, weekly context, "
    "and the trailing mean. Therefore a value that is globally normal can still "
    "look unusual when its lag/rolling context is unusual."
)


# ------------------------------------------------------------
# 21. CONCRETE CONTEXTUAL-ANOMALY EXAMPLE
# ------------------------------------------------------------

# Pick the highest multivariate KNN distance after excluding obvious extreme values.
context_examples = knn_multi[
    (knn_multi["knn_flag"]) &
    (knn_multi["sales"].abs() < 1000)
].copy()

if not context_examples.empty:
    example = context_examples.sort_values(
        "knn_avg_distance",
        ascending=False
    ).iloc[0]

    print("\nExercise 21 contextual example:")
    print(example[
        [
            "product_id",
            "date",
            "sales",
            "sales_lag_1",
            "sales_lag_7",
            "rolling_mean_7",
            "knn_avg_distance"
        ]
    ].to_string())

    print(
        "\nInterpretation: the raw sales value alone may not be globally extreme, "
        "but its recent/weekly context can make the feature vector unusual."
    )


# ------------------------------------------------------------
# 22. PRACTICAL TRADE-OFF
# ------------------------------------------------------------

tradeoff = pd.DataFrame(
    {
        "method": [
            "Z-score",
            "Modified Z-score",
            "Multivariate KNN"
        ],
        "main_choices": [
            "Per-product baseline + threshold (e.g. 3 SD)",
            "Per-product median/MAD + threshold (e.g. 3.5)",
            "k + feature set + scaling + top fraction/threshold"
        ],
        "plain_language_explanation": [
            "High/low compared with the product's global average.",
            "High/low compared with a robust median baseline.",
            "Different from nearby observations in the chosen feature space."
        ],
        "adding_product": [
            "Compute that product's baseline separately.",
            "Compute that product's median/MAD separately.",
            "Run the same feature pipeline and fit a model for the new product."
        ]
    }
)

print("\nExercise 22:")
print(tradeoff.to_string(index=False))

print(
    "\nSuggested practical conclusion:\n"
    "Use a simple Z-score or Modified Z-score as a transparent first-line monitor "
    "when global value anomalies are the main concern. Keep multivariate KNN for "
    "richer contextual/offline investigation unless its thresholding and false-positive "
    "behavior have been carefully validated on historical ground truth."
)
```

## EXERCISE 4 — GROUND TRUTH EVALUATION

```python
# EXERCISE 4 — GROUND TRUTH EVALUATION
```

```python
# ------------------------------------------------------------
# 23. LOAD anomaly_answer_key.csv
# ------------------------------------------------------------

KEY_PATH = DATA_DIR / "anomaly_answer_key.csv"

if not KEY_PATH.exists():
    print(
        "\nExercises 23–27 require anomaly_answer_key.csv.\n"
        "The supplied files currently contain daily_sales.csv and "
        "product_reference.csv, but the ground-truth key is not present.\n"
        "Place anomaly_answer_key.csv in the same folder and rerun this section."
    )
else:
    key = pd.read_csv(KEY_PATH)

    print("\nGround-truth key columns:")
    print(list(key.columns))

    print("\nGround-truth preview:")
    print(key.head().to_string(index=False))

    # --------------------------------------------------------
    # 24. Precision and Recall
    # --------------------------------------------------------
    #
    # The exact column names can differ by exercise version.
    # Adjust these three names if necessary.
    #
    # Expected logical fields:
    #   product_id
    #   date
    #   type
    #
    # Structural anomalies are deliberately not treated as value anomalies
    # because missing rows do not exist and duplicate rows are not inherently
    # numerical outliers.

    # Try common aliases automatically.
    def find_col(columns, candidates):
        lower = {c.lower(): c for c in columns}
        for candidate in candidates:
            if candidate.lower() in lower:
                return lower[candidate.lower()]
        return None

    key_product_col = find_col(
        key.columns,
        ["product_id", "product", "productid"]
    )
    key_date_col = find_col(
        key.columns,
        ["date", "anomaly_date", "timestamp"]
    )
    key_type_col = find_col(
        key.columns,
        ["type", "anomaly_type", "anomaly"]
    )

    if not all([key_product_col, key_date_col, key_type_col]):
        raise ValueError(
            "Could not automatically identify product/date/type columns "
            "in anomaly_answer_key.csv. Rename them or edit the column mapping."
        )

    key_eval = key.copy()
    key_eval["date"] = pd.to_datetime(key_eval[key_date_col])

    structural_types = {
        "missing_record",
        "duplicate_record"
    }

    value_key = key_eval[
        ~key_eval[key_type_col].astype(str).isin(structural_types)
    ].copy()

    # --------------------------------------------------------
    # Build detector flag tables
    # --------------------------------------------------------

    detector_tables = {
        "Z-score": z_df[
            ["product_id", "date", "z_flag"]
        ].rename(columns={"z_flag": "flag"}),

        "Modified Z-score": mz_df[
            ["product_id", "date", "mz_flag"]
        ].rename(columns={"mz_flag": "flag"}),

        "Multivariate KNN": knn_multi[
            ["product_id", "date", "knn_flag"]
        ].rename(columns={"knn_flag": "flag"})
    }

    # A practical evaluation key for existing rows.
    truth_pairs = set(
        zip(
            value_key[key_product_col].astype(str),
            value_key["date"]
        )
    )

    def evaluate_detector(detector_df):
        pred_pairs = set(
            zip(
                detector_df.loc[detector_df["flag"], "product_id"].astype(str),
                detector_df.loc[detector_df["flag"], "date"]
            )
        )

        tp = len(pred_pairs & truth_pairs)
        fp = len(pred_pairs - truth_pairs)
        fn = len(truth_pairs - pred_pairs)

        precision = tp / (tp + fp) if (tp + fp) else np.nan
        recall = tp / (tp + fn) if (tp + fn) else np.nan

        return tp, fp, fn, precision, recall

    evaluation_rows = []

    for method, detector_df in detector_tables.items():
        tp, fp, fn, precision, recall = evaluate_detector(detector_df)

        evaluation_rows.append(
            {
                "method": method,
                "TP": tp,
                "FP": fp,
                "FN": fn,
                "precision": precision,
                "recall": recall
            }
        )

    evaluation = pd.DataFrame(evaluation_rows)

    print("\nExercise 24 — overall evaluation:")
    print(evaluation.round(4).to_string(index=False))

    # --------------------------------------------------------
    # 25. Recall by anomaly type
    # --------------------------------------------------------

    anomaly_types = [
        "point_spike",
        "stockout_zero_run",
        "level_shift_start",
        "data_entry_error",
        "missing_record",
        "duplicate_record",
        "gradual_drift_start"
    ]

    recall_rows = []

    for method, detector_df in detector_tables.items():
        pred_pairs = set(
            zip(
                detector_df.loc[detector_df["flag"], "product_id"].astype(str),
                detector_df.loc[detector_df["flag"], "date"]
            )
        )

        row = {"method": method}

        for anomaly_type in anomaly_types:
            truth_type = key_eval[
                key_eval[key_type_col].astype(str) == anomaly_type
            ]

            # Structural anomalies cannot be represented by an ordinary
            # value-detector row. Report zero recall unless your exercise
            # defines a special structural matching rule.
            if anomaly_type in structural_types:
                row[anomaly_type] = 0.0
                continue

            truth_type_pairs = set(
                zip(
                    truth_type[key_product_col].astype(str),
                    truth_type["date"]
                )
            )

            tp = len(pred_pairs & truth_type_pairs)
            denom = len(truth_type_pairs)

            row[anomaly_type] = tp / denom if denom else np.nan

        recall_rows.append(row)

    recall_by_type = pd.DataFrame(recall_rows)

    print("\nExercise 25 — recall by anomaly type:")
    print(recall_by_type.round(3).to_string(index=False))

    # --------------------------------------------------------
    # 26. Structural anomaly confirmation
    # --------------------------------------------------------

    print(
        "\nExercise 26:\n"
        "missing_record and duplicate_record are structural anomalies. "
        "A missing row has no sales value on which to calculate a score, and "
        "a duplicate can be numerically ordinary. Therefore failure to detect "
        "them with these three value/distance detectors is expected."
    )

    # --------------------------------------------------------
    # 27. Rank methods by recall and precision
    # --------------------------------------------------------

    target_types = [
        "point_spike",
        "stockout_zero_run",
        "level_shift_start",
        "data_entry_error",
        "gradual_drift_start"
    ]

    ranking_source = recall_by_type[
        ["method"] + target_types
    ].copy()

    ranking_source["mean_recall_target_types"] = (
        ranking_source[target_types].mean(axis=1)
    )

    recall_rank = (
        ranking_source
        .sort_values("mean_recall_target_types", ascending=False)
        [["method", "mean_recall_target_types"]]
    )

    precision_rank = (
        evaluation
        .sort_values("precision", ascending=False)
        [["method", "precision"]]
    )

    print("\nExercise 27 — recall ranking:")
    print(recall_rank.round(4).to_string(index=False))

    print("\nExercise 27 — precision ranking:")
    print(precision_rank.round(4).to_string(index=False))

    print(
        "\nTrade-off interpretation:\n"
        "Recall rewards catching more true anomalies. Precision rewards avoiding "
        "false alarms. A method can therefore rank first on recall but lower on "
        "precision if it flags many borderline points."
    )
```
