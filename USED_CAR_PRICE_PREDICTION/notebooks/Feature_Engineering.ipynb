# Objectives

   - Load the cleaned dataset
   - Separate features and target
   - Identify categorical and numerical columns
   - Build a preprocessing pipeline
   - Encode categorical variables
   - Scale numerical variables
   - Split the dataset
   - Save the preprocessing pipeline

```python
import warnings
warnings.filterwarnings("ignore")

import joblib
import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)

from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("../data/processed/used_cars_clean.csv")
df.head()
```

```python
print(df.shape)
```

```python
X = df.drop("price", axis=1)
y = df["price"]
```

```python
# Display feature name
print("Input Features\n")
for feature in X.columns:
    print(feature)
```

```python
# Identify Numerical column
numeric_features = X.select_dtypes(
    include=np.number
).columns.tolist()

numeric_features
```

```python
# Identify Categorical column
categorical_features = X.select_dtypes(
    include="object").columns.tolist()

categorical_features
```

```python
# Numerical Pipeline
numeric_pipeline = Pipeline(
    steps=[
        ("imputer",SimpleImputer(strategy="median")),
         ("scaler",StandardScaler())])
```

```python
# Categorical Pipeline
categorical_pipeline = Pipeline(
    steps=[
        ("imputer",SimpleImputer(strategy="most_frequent")),
        ("encoder",OneHotEncoder(handle_unknown="ignore"))])
```

```python
# Create Preprocessor
preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_pipeline,
            numeric_features
        ),

        (
            "cat",
            categorical_pipeline,
            categorical_features
        )
    ]
)
```

```python
# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
print("Training Samples :", X_train.shape)

print("Testing Samples :", X_test.shape)
```

```python
# Fit Prerocessor
preprocessor.fit(X_train)
```

```python
# Transform the data
X_train_processed = preprocessor.transform(X_train)
X_test_processed = preprocessor.transform(X_test)
```

```python
# check the shape
print(X_train_processed.shape)
print(X_test_processed.shape)
```

```python
# Save preprocessing Pipeline
joblib.dump(
    preprocessor,
    "../models/preprocessing/preprocessor.pkl"
)
```

```python
# Save Feature Lists
joblib.dump(
    numeric_features,
    "../models/metadata/numeric_features.pkl"
)

joblib.dump(
    categorical_features,
    "../models/metadata/categorical_features.pkl"
)
```

```python
# Save Train/Test Data

joblib.dump(
    X_train_processed,
    "../models/preprocessing/X_train.pkl"
)

joblib.dump(
    X_test_processed,
    "../models/preprocessing/X_test.pkl"
)

joblib.dump(
    y_train,
    "../models/preprocessing/y_train.pkl"
)

joblib.dump(
    y_test,
    "../models/preprocessing/y_test.pkl"
)
```

```python
# Verified the saved file
import os
for root, _, files in os.walk("../models"):
    for file in files:
        print(os.path.join(root, file))
```
