```python
# ============================================================
# IMPORT LIBRARIES
# ============================================================

import os
import random
import shutil
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from PIL import Image

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

print("TensorFlow version:", tf.__version__)
```

```python
# ============================================================
# PROJECT CONFIGURATION
# ============================================================

DATASET_ROOT = Path("chest_xray_data/chest_xray")

TRAIN_DIR = DATASET_ROOT / "train"
TEST_DIR = DATASET_ROOT / "test"
VAL_DIR = DATASET_ROOT / "val"

CLASS_NAMES = ["NORMAL", "PNEUMONIA"]

IMG_HEIGHT = 150
IMG_WIDTH = 150
IMG_SIZE = (IMG_HEIGHT, IMG_WIDTH)

BATCH_SIZE = 32
SEED = 42

VALIDATION_SPLIT = 0.15

EPOCHS = 30

print("Dataset root:", DATASET_ROOT)
print("Train directory:", TRAIN_DIR)
print("Test directory:", TEST_DIR)
```

```python
# ============================================================
# VERIFY DATASET STRUCTURE
# ============================================================

required_directories = [
    TRAIN_DIR / "NORMAL",
    TRAIN_DIR / "PNEUMONIA",
    TEST_DIR / "NORMAL",
    TEST_DIR / "PNEUMONIA",
    VAL_DIR / "NORMAL",
    VAL_DIR / "PNEUMONIA"
]

for directory in required_directories:
    print(f"{directory}: {'FOUND' if directory.exists() else 'NOT FOUND'}")
```

```python
# Task 1 — Inspect Before You Touch Anything
# ============================================================
# TASK 1.1 — COUNT IMAGES PER CLASS
# ============================================================

IMAGE_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".gif",
    ".tif",
    ".tiff"
}


def count_images(directory):
    counts = {}

    for class_name in CLASS_NAMES:
        class_dir = directory / class_name

        if not class_dir.exists():
            counts[class_name] = 0
            continue

        image_files = [
            file
            for file in class_dir.iterdir()
            if file.is_file() and file.suffix.lower() in IMAGE_EXTENSIONS
        ]

        counts[class_name] = len(image_files)

    return counts


train_counts = count_images(TRAIN_DIR)
test_counts = count_images(TEST_DIR)
val_counts = count_images(VAL_DIR)

print("TRAIN:")
for class_name, count in train_counts.items():
    print(f"  {class_name}: {count:,}")

print("\nVALIDATION PROVIDED BY DATASET:")
for class_name, count in val_counts.items():
    print(f"  {class_name}: {count:,}")

print("\nTEST:")
for class_name, count in test_counts.items():
    print(f"  {class_name}: {count:,}")
```

```python
# ============================================================
# CLASS IMBALANCE RATIO
# ============================================================

normal_count = train_counts["NORMAL"]
pneumonia_count = train_counts["PNEUMONIA"]

imbalance_ratio = pneumonia_count / normal_count

print(f"NORMAL images:    {normal_count:,}")
print(f"PNEUMONIA images: {pneumonia_count:,}")
print(f"\nPNEUMONIA : NORMAL = {imbalance_ratio:.2f} : 1")
```

```python
# ============================================================
# VISUALIZE CLASS DISTRIBUTION
# ============================================================

plt.figure(figsize=(7, 5))

plt.bar(
    train_counts.keys(),
    train_counts.values()
)

plt.title("Training Dataset Class Distribution")
plt.xlabel("Class")
plt.ylabel("Number of Images")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# GET IMAGE FILES
# ============================================================

def get_image_files(directory, class_name):
    class_dir = directory / class_name

    return [
        file
        for file in class_dir.iterdir()
        if file.is_file() and file.suffix.lower() in IMAGE_EXTENSIONS
    ]


normal_files = get_image_files(TRAIN_DIR, "NORMAL")
pneumonia_files = get_image_files(TRAIN_DIR, "PNEUMONIA")

print("NORMAL:", len(normal_files))
print("PNEUMONIA:", len(pneumonia_files))
```

```python
# ============================================================
# DISPLAY 5 IMAGES PER CLASS
# ============================================================

random.seed(SEED)
sample_normal = random.sample(normal_files,min(5, len(normal_files)))
sample_pneumonia = random.sample(pneumonia_files,min(5, len(pneumonia_files)))
fig, axes = plt.subplots(2,5,figsize=(18, 7))

# NORMAL
for i, image_path in enumerate(sample_normal):
    image = Image.open(image_path)

    axes[0, i].imshow(image, cmap="gray")
    axes[0, i].set_title("NORMAL")
    axes[0, i].axis("off")


# PNEUMONIA
for i, image_path in enumerate(sample_pneumonia):
    image = Image.open(image_path)

    axes[1, i].imshow(image, cmap="gray")
    axes[1, i].set_title("PNEUMONIA")
    axes[1, i].axis("off")

plt.suptitle("Sample Chest X-Ray Images",fontsize=16)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# CHECK IMAGE DIMENSIONS AND COLOR MODES
# ============================================================

def inspect_images(directory, max_images=None):
    records = []
    for class_name in CLASS_NAMES:
        files = get_image_files(directory, class_name)
        if max_images is not None:
            files = files[:max_images]
        for image_path in files:
            try:
                with Image.open(image_path) as image:
                    records.append({
                        "class": class_name,
                        "filename": image_path.name,
                        "width": image.width,
                        "height": image.height,
                        "mode": image.mode
                    })
            except Exception as error:
                print(f"Could not read {image_path}: {error}")
    return pd.DataFrame(records)

inspection_df = inspect_images(
    TRAIN_DIR,
    max_images=None)
inspection_df.head()
```

```python
# ============================================================
# IMAGE DIMENSION SUMMARY
# ============================================================

print("Unique image dimensions:")

dimension_counts = (
    inspection_df
    .groupby(["width", "height"])
    .size()
    .sort_values(ascending=False)
)

print(dimension_counts.head(20))

print("\nNumber of unique dimensions:")
print(
    inspection_df[["width", "height"]]
    .drop_duplicates()
    .shape[0]
)
```

```python
# ============================================================
# COLOR MODE SUMMARY
# ============================================================

mode_counts = inspection_df["mode"].value_counts()

print("Image color modes:")
print(mode_counts)
```

```python
# Task 2 — Build a Proper Train/Validation Split
# ============================================================
# TASK 2 — CREATE STRATIFIED TRAIN/VALIDATION SPLIT
# ============================================================

all_files = []
all_labels = []

for class_index, class_name in enumerate(CLASS_NAMES):

    files = get_image_files(
        TRAIN_DIR,
        class_name
    )

    for file in files:

        all_files.append(str(file))
        all_labels.append(class_index)


all_files = np.array(all_files)
all_labels = np.array(all_labels)

print("Total images:", len(all_files))
print("Total labels:", len(all_labels))
```

```python
# ============================================================
# STRATIFIED 85/15 SPLIT
# ============================================================

train_files, val_files, train_labels, val_labels = train_test_split(
    all_files,
    all_labels,
    test_size=VALIDATION_SPLIT,
    random_state=SEED,
    stratify=all_labels
)

print("Training images:", len(train_files))
print("Validation images:", len(val_files))
```

```python
# ============================================================
# VERIFY CLASS DISTRIBUTION AFTER SPLIT
# ============================================================

def show_split_distribution(labels, split_name):

    values, counts = np.unique(
        labels,
        return_counts=True
    )

    print(f"\n{split_name}")

    for value, count in zip(values, counts):

        print(
            f"{CLASS_NAMES[value]}: {count:,}"
        )


show_split_distribution(
    train_labels,
    "TRAINING SPLIT"
)

show_split_distribution(
    val_labels,
    "VALIDATION SPLIT"
)
```

```python
# ============================================================
# CLASS PERCENTAGES
# ============================================================

train_distribution = pd.Series(
    train_labels
).map(
    dict(enumerate(CLASS_NAMES))
).value_counts()

val_distribution = pd.Series(
    val_labels
).map(
    dict(enumerate(CLASS_NAMES))
).value_counts()

distribution_df = pd.DataFrame({
    "Training": train_distribution,
    "Validation": val_distribution
})

distribution_df["Training %"] = (
    distribution_df["Training"]
    / distribution_df["Training"].sum()
    * 100
)

distribution_df["Validation %"] = (
    distribution_df["Validation"]
    / distribution_df["Validation"].sum()
    * 100
)

distribution_df
```

```python
# Task 3 — Preprocessing Pipeline
# ============================================================
# CREATE TF.DATA DATASETS
# ============================================================

AUTOTUNE = tf.data.AUTOTUNE


def load_and_preprocess_image(file_path, label):

    image = tf.io.read_file(file_path)
    image = tf.image.decode_image(image,channels=1,expand_animations=False)
    image.set_shape([None, None, 1])
    image = tf.image.resize(image,IMG_SIZE)
    image = tf.cast(image,tf.float32) / 255.0
    return image, label

train_ds = tf.data.Dataset.from_tensor_slices((train_files, train_labels))
val_ds = tf.data.Dataset.from_tensor_slices((val_files, val_labels))

train_ds = (train_ds.shuffle(buffer_size=len(train_files),seed=SEED)
    .map(load_and_preprocess_image,num_parallel_calls=AUTOTUNE)
    .batch(BATCH_SIZE)
    .prefetch(AUTOTUNE))

val_ds = (val_ds
    .map(load_and_preprocess_image,num_parallel_calls=AUTOTUNE)
    .batch(BATCH_SIZE)
    .prefetch(AUTOTUNE))
```

```python
# ============================================================
# DATA AUGMENTATION
# ============================================================

data_augmentation = keras.Sequential(
    [
        layers.RandomRotation(factor=0.04),
        layers.RandomZoom(height_factor=0.10,width_factor=0.10),
        layers.RandomFlip(mode="horizontal"),
        layers.RandomBrightness(factor=0.10)],
    name="data_augmentation"
)
```

```python
# ============================================================
# VISUALIZE DATA AUGMENTATION
# ============================================================

sample_images, sample_labels = next(iter(train_ds))
sample_image = sample_images[0]
plt.figure(figsize=(12, 8))

for i in range(6):
    augmented_image = data_augmentation(tf.expand_dims(sample_image, axis=0),training=True)
    augmented_image = augmented_image[0]
    plt.subplot(2, 3, i + 1)
    plt.imshow(augmented_image,cmap="gray")
    plt.axis("off")
    plt.title("Augmented")

plt.suptitle("Training Data Augmentation")
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# COMPUTE CLASS WEIGHTS
# ============================================================

classes = np.unique(train_labels)
class_weights_array = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=train_labels)

class_weights = {
    int(class_index): float(weight)
    for class_index, weight
    in zip(classes, class_weights_array)
}

print("Class weights:")

for class_index, weight in class_weights.items():
    print(f"{CLASS_NAMES[class_index]}: {weight:.4f}")
```

```python
# Task 4 — Build the CNN
# ============================================================
# TASK 4 — CNN FROM SCRATCH
# ============================================================

def build_cnn():

    model = keras.Sequential(
        [
            # ------------------------------------------------
            # INPUT
            # ------------------------------------------------
            layers.Input(shape=(IMG_HEIGHT, IMG_WIDTH, 1)),

            # ------------------------------------------------
            # DATA AUGMENTATION
            # ------------------------------------------------
            data_augmentation,

            # ------------------------------------------------
            # BLOCK 1
            # ------------------------------------------------
            layers.Conv2D(32,(3, 3),padding="same"),
            layers.BatchNormalization(),
            layers.ReLU(),
            layers.MaxPooling2D(pool_size=(2, 2)),

            # ------------------------------------------------
            # BLOCK 2
            # ------------------------------------------------
            layers.Conv2D(64,(3, 3),padding="same"),
            layers.BatchNormalization(),
            layers.ReLU(),
            layers.MaxPooling2D(pool_size=(2, 2)),
            layers.Dropout(0.25),

            # ------------------------------------------------
            # BLOCK 3
            # ------------------------------------------------
            layers.Conv2D(128,(3, 3),padding="same"),
            layers.BatchNormalization(),
            layers.ReLU(),
            layers.MaxPooling2D(pool_size=(2, 2)),
            layers.Dropout(0.30),

            # ------------------------------------------------
            # BLOCK 4
            # ------------------------------------------------
            layers.Conv2D(256,(3, 3),padding="same"),
            layers.BatchNormalization(),
            layers.ReLU(),
            layers.MaxPooling2D(pool_size=(2, 2)),
            layers.Dropout(0.35),

            # ------------------------------------------------
            # CLASSIFICATION HEAD
            # ------------------------------------------------
            layers.GlobalAveragePooling2D(),
            layers.Dense(128,activation="relu"),
            layers.Dropout(0.50),

            # Binary classification
            layers.Dense(1,activation="sigmoid")],
        name="Chest_XRay_CNN")

    return model
```

```python
# ============================================================
# CREATE MODEL
# ============================================================

model = build_cnn()

model.summary()
```

```python
# ============================================================
# COMPILE MODEL
# ============================================================

model.compile(optimizer=keras.optimizers.Adam(learning_rate=1e-3),

    loss="binary_crossentropy",
    metrics=[keras.metrics.BinaryAccuracy(name="accuracy"),
        keras.metrics.Precision(name="precision"),
        keras.metrics.Recall(name="recall"),
        keras.metrics.AUC(name="auc")]
)
```

```python
# Task 5 — Train With Class Weights
# ============================================================
# TRAINING CALLBACKS
# ============================================================

early_stopping = keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True,
    verbose=1
)


reduce_lr = keras.callbacks.ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=2,
    min_lr=1e-6,
    verbose=1
)


checkpoint = keras.callbacks.ModelCheckpoint(
    "best_chest_xray_cnn.keras",
    monitor="val_loss",
    save_best_only=True,
    verbose=1
)
```

```python
# ============================================================
# TRAIN CNN WITH CLASS WEIGHTS
# ============================================================

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS,
    class_weight=class_weights,
    callbacks=[early_stopping,reduce_lr,checkpoint],
    verbose=1
)
```

```python
# Plot Training Accuracy
# ============================================================
# TRAINING / VALIDATION ACCURACY
# ============================================================

history_df = pd.DataFrame(history.history)
plt.figure(figsize=(9, 6))
plt.plot(history_df["accuracy"],label="Training Accuracy")
plt.plot(history_df["val_accuracy"],label="Validation Accuracy")
plt.title("Training and Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Plot Training L
# ============================================================
# TRAINING / VALIDATION LOSS
# ============================================================

plt.figure(figsize=(9, 6))
plt.plot(history_df["loss"],label="Training Loss")
plt.plot(history_df["val_loss"],label="Validation Loss")
plt.title("Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Binary Cross-Entropy Loss")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Task 6 — Evaluate Properly
# ============================================================
# CREATE TEST DATASET
# ============================================================

test_files = []
test_labels = []

for class_index, class_name in enumerate(CLASS_NAMES):
    files = get_image_files(TEST_DIR,class_name)
    
    for file in files:
        test_files.append(str(file))
        test_labels.append(class_index)

test_files = np.array(test_files)
test_labels = np.array(test_labels)
test_ds = tf.data.Dataset.from_tensor_slices((test_files, test_labels))
test_ds = (test_ds
    .map(load_and_preprocess_image,num_parallel_calls=AUTOTUNE)
    .batch(BATCH_SIZE)
    .prefetch(AUTOTUNE))

print("Test images:", len(test_files))
```

```python
# ============================================================
# TEST SET EVALUATION
# ============================================================

test_results = model.evaluate(test_ds,verbose=1)
print("\nTest Results:")

for metric_name, metric_value in zip(model.metrics_names,test_results):
    print(f"{metric_name}: {metric_value:.4f}")
```

```python
# ============================================================
# GENERATE TEST PREDICTIONS
# ============================================================

y_probability = model.predict(test_ds,verbose=1).ravel()
y_pred = (y_probability >= 0.5).astype(int)
y_true = test_labels

print("Predictions generated:", len(y_pred))
```

```python
# ============================================================
# CLASSIFICATION REPORT
# ============================================================

print(
    classification_report(
        y_true,y_pred,
        target_names=CLASS_NAMES,digits=4
    ))
```

```python
# ============================================================
# ACCURACY
# ============================================================

accuracy = accuracy_score(y_true,y_pred)
print(f"Test Accuracy: {accuracy:.4f}")
```

```python
# ============================================================
# PRECISION
# ============================================================

precision = precision_score(y_true,y_pred,average=None,zero_division=0)

for class_name, value in zip(CLASS_NAMES,precision):
    print(f"{class_name} Precision: {value:.4f}")
```

```python
# ============================================================
# RECALL
# ============================================================

recall = recall_score(y_true,y_pred,average=None,zero_division=0)

for class_name, value in zip(CLASS_NAMES,recall):
    print(f"{class_name} Recall: {value:.4f}")
```

```python
normal_recall = recall[0]

print(f"NORMAL Recall: {normal_recall:.4f}")
```

```python
# ============================================================
# F1 SCORE
# ============================================================

f1 = f1_score(y_true,y_pred,average=None,zero_division=0)
for class_name, value in zip(CLASS_NAMES,f1):
    print(f"{class_name} F1-score: {value:.4f}")
```

```python
# ============================================================
# ROC-AUC
# ============================================================

auc_score = roc_auc_score(y_true,y_probability)
print(f"Test ROC-AUC: {auc_score:.4f}")
```

```python
# ============================================================
# CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(y_true,y_pred)
print(cm)

# ============================================================
# CONFUSION MATRIX VISUALIZATION
# ============================================================

plt.figure(figsize=(7, 6))
plt.imshow(cm)
plt.title("Confusion Matrix — Chest X-Ray CNN")
plt.xlabel("Predicted Label")
plt.ylabel("True Label")

plt.xticks([0, 1],CLASS_NAMES)
plt.yticks([0, 1],CLASS_NAMES)

for i in range(2):
    for j in range(2):
        plt.text(j,i,cm[i, j],ha="center",va="center",fontsize=14)

plt.colorbar()
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# CONFUSION MATRIX VALUES
# ============================================================

tn, fp, fn, tp = cm.ravel()

print("True Negative :", tn)
print("False Positive:", fp)
print("False Negative:", fn)
print("True Positive :", tp)
```

```python
# ============================================================
# PNEUMONIA FALSE NEGATIVE RATE
# ============================================================

pneumonia_recall = tp / (tp + fn)
pneumonia_false_negative_rate = (fn / (tp + fn))

print(f"PNEUMONIA Recall: {pneumonia_recall:.4f}")
print(
    f"PNEUMONIA False Negative Rate: "
    f"{pneumonia_false_negative_rate:.4f}")
```

```python
# 7 Build Unweighted Model
# ============================================================
# BONUS — MODEL WITHOUT CLASS WEIGHTS
# ============================================================

unweighted_model = build_cnn()
unweighted_model.compile(optimizer=keras.optimizers.Adam(learning_rate=1e-3),

    loss="binary_crossentropy",
    metrics=[
        keras.metrics.BinaryAccuracy(name="accuracy"),
        keras.metrics.Precision(name="precision"),
        keras.metrics.Recall(name="recall"),
        keras.metrics.AUC(name="auc")
    ]
)
```

```python
# ============================================================
# TRAIN WITHOUT CLASS WEIGHTS
# ============================================================

unweighted_history = unweighted_model.fit(
    train_ds,

    validation_data=val_ds,

    epochs=EPOCHS,

    callbacks=[
        keras.callbacks.EarlyStopping(
            monitor="val_loss",
            patience=5,
            restore_best_weights=True,
            verbose=1
        ),

        keras.callbacks.ReduceLROnPlateau(
            monitor="val_loss",
            factor=0.5,
            patience=2,
            min_lr=1e-6,
            verbose=1
        )
    ],

    verbose=1
)
```

```python
# ============================================================
# PREDICTIONS WITHOUT CLASS WEIGHTS
# ============================================================

unweighted_probability = (
    unweighted_model
    .predict(test_ds, verbose=1)
    .ravel()
)

unweighted_pred = (
    unweighted_probability >= 0.5
).astype(int)
```

```python
# ============================================================
# COMPARE WEIGHTED VS UNWEIGHTED
# ============================================================

print("=" * 70)
print("MODEL WITH CLASS WEIGHTS")
print("=" * 70)

print(
    classification_report(
        y_true,
        y_pred,
        target_names=CLASS_NAMES,
        digits=4
    )
)


print("=" * 70)
print("MODEL WITHOUT CLASS WEIGHTS")
print("=" * 70)

print(
    classification_report(
        y_true,
        unweighted_pred,
        target_names=CLASS_NAMES,
        digits=4
    )
)
```

```python
# ============================================================
# SINGLE IMAGE PREDICTION
# ============================================================

def predict_xray(image_path, model):

    image = tf.io.read_file(str(image_path))
    image = tf.image.decode_image(image,channels=1,expand_animations=False)
    image.set_shape([None, None, 1])
    image = tf.image.resize(image,IMG_SIZE)
    image = tf.cast(image,tf.float32) / 255.0
    image_batch = tf.expand_dims(image,axis=0)
    probability = model.predict(image_batch,verbose=0)[0][0]

    if probability >= 0.5:
        
        predicted_class = "PNEUMONIA"
        confidence = probability

    else:

        predicted_class = "NORMAL"
        confidence = 1 - probability

    return predicted_class, float(confidence)
```

```python
# ============================================================
# EXAMPLE PREDICTION
# ============================================================

sample_path = test_files[0]

prediction, confidence = predict_xray(sample_path,model)

print("Image:", sample_path)
print("Prediction:", prediction)
print(f"Confidence: {confidence:.2%}")
```
