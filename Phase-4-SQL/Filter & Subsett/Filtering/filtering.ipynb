```python
import pandas as pd
data = {
 'Order_ID': ['1001', '1002', '1003', '1004', '1005'],
 'Product_ID': ['P101', 'P102', 'P103', 'P101', 'P104'],
 'Product_Name': ['Laptop', 'Smartphone', 'Headphones', 'Laptop',
'Coffee Maker'],
 'Category': ['Electronics', 'Electronics', 'Accessories',
'Electronics', 'Kitchen'],
 'Quantity': [2, 1, 5, 3, 2],
 'Amount': [1600, 600, 500, 2400, 100],
 'Date': ['2024-01-10', '2024-01-11', '2024-01-12', '2024-01-12',
'2024-01-13'],
 'Region': ['North', 'South', 'East', 'West', 'North'],
}

filtered=pd.DataFrame(data)
filtered
```

```python
print("Fitering :- is the process of selecting specific aount of data from the given dataset")
```

```python
print("To filter only sales related to Electronics products")
Category=filtered[filtered["Category"]=="Electronics"]
Category
```

```python
print(" To filter sales that are greater than 1,000")
Sales=filtered[filtered["Amount"]>1000]
Sales
```

```python
print("Filtered by Total Amount (>1000) AND Date (Before Jan 12, 2024)")
filtered["Date"]=pd.to_datetime(filtered["Date"])
Filter=filtered[(filtered["Amount"]>1000) & (filtered["Date"]<'2024-01-12')]
Filter
```
