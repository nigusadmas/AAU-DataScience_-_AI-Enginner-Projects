```python
import pandas as pd
data = {
 'Order_ID': ['1001', '1002', '1003', '1004', '1005'],
 'Product_ID': ['P101', 'P102', 'P103', 'P101', 'P104'],
 'Product_Name': ['Laptop', 'Smartphone', 'Headphones', 'Laptop',
'Coffee Maker'],
 'Category': ['Electronics', 'Electronics', 'Accessories',
'Electronics', 'Kitchen'],
 'Quantity': [2, 1, 5, 3, 2],
 'Amount': [1600, 600, 500, 2400, 100],
 'Date': ['2024-01-10', '2024-01-11', '2024-01-12', '2024-01-12',
'2024-01-13'],
 'Region': ['North', 'South', 'East', 'West', 'North'],
}
df_sales=pd.DataFrame(data)
df_sales
```

```python
print("Subsetting ማለት ከdataset ውስጥ የሚያስፈልጉትን rows, columns, ወይም ሁለቱንም መምረጥ ነው።")
```

```python
# Subsetting only Order_ID,product_name and amount
subset_df = df_sales[['Order_ID', 'Product_Name', 'Amount']]
subset_df
```

```python
print("We can use iloc function(method) to filter rows in subsetting")
Row=df_sales.iloc[:3]
Row
```
