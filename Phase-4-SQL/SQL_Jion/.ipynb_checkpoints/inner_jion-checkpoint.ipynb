```python
# Dataset 1: Customers
import pandas as pd
customers = pd.DataFrame({
"Customer_ID": ["C001", "C002", "C003", "C004", "C005", "C006"],
"Customer_Name": ["Abel", "Sara", "John", "Helen", "Samuel","David"],
"City": ["Addis Ababa", "Adama", "Jimma", "Bahir Dar", "Hawassa","Dire Dawa"]
})
customers
```

```python
# Dataset 2: Orders
orders = pd.DataFrame({
"Order_ID": [101,102,103,104,105,106,107],
"Customer_ID": ["C001","C002","C001","C004","C007","C006","C006"],"Product":
["Laptop","Phone","Mouse","Keyboard","Tablet","Monitor","Printer"],"Amount": [1200,800,45,150,650,700,300]
})
orders
```

```python
# Exercise 1 — Customer Purchase Report
# The sales manager wants a report that contains only valid purchases made by registeredcustomers.
# Create a report showing:
# Customer Name,City,Product,Amount
# Questions
# 1. How many records are returned?
# Total record returnd=6
# 2. Which order is excluded from the report?
# Order_ID = 105
# 3. Why was that order excluded?
# It belongs to Customer_ID = C007,But C007 is NOT in customers table
report=pd.merge(customers,orders,on="Customer_ID" , how="inner")
report = report[["Customer_Name", "City", "Product", "Amount"]]
report
```

```python
# Exercise 2 — Order Investigation
# The finance department wants to review every order, even if customer information is
# unavailable.
# Create a report showing:
# Order ID
# Customer ID
# Customer Name
# Product
# Amount
# Questions
# 1. Which order has missing customer information?
# order_Id=105
# 2. Why is the customer information missing?
# Orders table has C007
# But customers table does NOT contain C007
# So join cannot find matching row
# 3. How many registered customers do not appear in this report?
# Answer: 0 customers
report2 = pd.merge(customers, orders, on="Customer_ID", how="right")
report2 = report2[["Order_ID","Customer_ID","Customer_Name","Product","Amount"]]
report2
```

```python
# Exercise 3 — Customer Engagement
# Report
# The marketing team wants a list of all registered customers, including those who have
# never purchased anything.
# Display:
# Customer Name
# City
# Product
# Amount
# Questions
# 1. Which customer has never placed an order?
# 2. What values appear for that customer's Product and Amount?
# 3. Why do those values appear?


report3 = pd.merge(customers, orders, on="Customer_ID", how="left")
report3 = report3[["Customer_Name","City","Product","Amount"]]
report3
```

```python
# Exercise 4 — Data Quality Audit
# The data engineering team wants to compare both datasets and identify any mismatches.
# Create a report that includes every customer record and every order record.
# Questions
# 1. How many rows are returned?
# 2. Which registered customer has no orders?
# 3. Which order belongs to a customer that is not registered?
# 4. Why are both records important to keep during a data quality audit?

report4 = pd.merge(customers, orders, on="Customer_ID", how="outer")
report4
```

```python
# Exercise 5 — Customer Support
# Investigation
# Customer David contacts customer support claiming he has placed several orders.
# Generate a report showing:
# Customer Name
# Product
# Amount
# Questions
# 1. How many orders has David placed?
# 2. What is his total spending?
# 3. Would this report include customers who have never placed orders? Explain your answer.

df = pd.merge(customers, orders, on="Customer_ID", how="inner")

david = df[df["Customer_Name"] == "David"]

david
```
