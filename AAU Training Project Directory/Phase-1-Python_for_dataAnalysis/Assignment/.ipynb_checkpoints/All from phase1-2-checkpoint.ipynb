```python
import matplotlib.pyplot as plt
import numpy as np
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temperature = [22, 24, 23, 26, 25, 23, 24]
plt.plot(days,temperature, marker="o" ,markersize=8 , linewidth=4)
plt.title('Weekly Temperature')
plt.xlabel('Days')
plt.ylabel('Temperature (°C)')
plt.show()
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots()
fruits = ['apple', 'blueberry', 'Avocado', 'orange']
counts = [40, 75, 30, 55]
bar_labels = ['red', 'blue', 'green', 'orange']
bar_colors = ['tab:red', 'tab:blue', 'tab:green', 'tab:orange']
ax.bar(fruits, counts, label=bar_labels, color=bar_colors)
ax.set_ylabel('Supply')
ax.set_title('Fruit by kind and color')
ax.legend(title='Color')
plt.show()
```

```python
import numpy as np
dept = ('CS', 'Biology', 'IS', 'CE')
sex_counts = {'Male': np.array([73, 34, 61, 56]), 
              'Female': np.array([57, 34, 58,29])}
width = 0.75
fig, ax = plt.subplots()
bottom = np.zeros(4)
for sex, sex_count in sex_counts.items():
 p = ax.bar(dept, sex_count, width, label=sex, bottom=bottom)
 bottom += sex_count
 ax.bar_label(p, label_type='center')
ax.set_title('Number of student by sex for different courses')
ax.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

marks = [45,48,50,52,55,60,62,65,67,
         70,72,75,78,80,82,85,88,90,92,95]

plt.hist(marks,bins=10)

plt.title("Students Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
```

```python
# Import seaborn and apply the default theme
import seaborn as sns
sns.set_theme()
tips=sns.load_dataset("tips")
sns.relplot(data=tips, x="total_bill", y="tip", col="sex", hue="smoker", style="smoker", size="size",)
```

```python
fmri = sns.load_dataset("fmri")
sns.relplot(
 data=fmri, kind="line", x="timepoint", y="signal", col="region",
 hue="event", style="event",
)
```
