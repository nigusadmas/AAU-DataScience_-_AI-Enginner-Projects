```python
# ===========================================================
# Exercise 9
# Create a Dash application with:
# 1. A page title
# 2. A Dropdown for selecting movie genre
# 3. A Plotly graph
# 4. Display average movie score by rating
# ===========================================================
import pandas as pd
import plotly.express as px
from dash import Dash, dcc, html, Input, Output

# --------------------------------------------------
# Load Dataset
# --------------------------------------------------
movies = pd.read_csv("movies_updated.csv")

# --------------------------------------------------
# Create Dash App
# --------------------------------------------------
app = Dash(__name__)
app.title = "Movie Rating Dashboard"
# --------------------------------------------------
# Layout
# --------------------------------------------------
app.layout = html.Div([
    html.H1(
        "Movie Score Dashboard",
        style={"textAlign": "center"}
    ),

    html.Label("Select Movie Genre"),

    dcc.Dropdown(
        id="genre-dropdown",
        options=[
            {"label": genre, "value": genre}
            for genre in sorted(movies["genre"].unique())
        ],
        value=sorted(movies["genre"].unique())[0],
        clearable=False
        style={"width": "50%"}
    ),

    html.Br(),

    dcc.Graph(id="score-graph")
])

# --------------------------------------------------
# Callback
# --------------------------------------------------
@app.callback(
    Output("score-graph", "figure"),
    Input("genre-dropdown", "value")
)
def update_graph(selected_genre):

    filtered = movies[movies["genre"] == selected_genre]

    avg_score = (
        filtered.groupby("rating")["score"]
        .mean()
        .reset_index()
    )

    fig = px.bar(
        avg_score,
        x="rating",
        y="score",
        color="rating",
        text="score",
        title=f"Average Movie Score by Rating ({selected_genre})"
    )

    fig.update_traces(
        texttemplate="%{text:.2f}",
        textposition="outside"
    )

    fig.update_layout(
        xaxis_title="Movie Rating",
        yaxis_title="Average Score",
        template="plotly_white"
    )

    return fig

# --------------------------------------------------
# Run App
# --------------------------------------------------
if __name__ == "__main__":
    app.run(debug=True, port=8051)
```

```python
# Exercise 12
# Add a Year Range Slider to the dashboard.
# Requirements:
# Allow users to select a range of years.
# Update all graphs automatically based on the selected year range.

import pandas as pd
import plotly.express as px
from dash import Dash, dcc, html, Input, Output

# --------------------------------------------------
# Load dataset
# --------------------------------------------------
movies = pd.read_csv("movies_updated.csv")

# Ensure Year column is numeric
movies["year"] = pd.to_numeric(movies["year"], errors="coerce")

# --------------------------------------------------
# Create app
# --------------------------------------------------
app = Dash(__name__)
app.title = "Movie Dashboard"

# Get min/max year
min_year = int(movies["year"].min())
max_year = int(movies["year"].max())

# --------------------------------------------------
# Layout
# --------------------------------------------------
app.layout = html.Div([

    html.H1("Movie Dashboard", style={"textAlign": "center"}),

    # ---------------- Year Range Slider ----------------
    html.Label("Select Year Range"),

    dcc.RangeSlider(
        id="year-slider",
        min=min_year,
        max=max_year,
        value=[min_year, max_year],
        marks={str(year): str(year) for year in range(min_year, max_year + 1, 5)},
        step=1
    ),

    html.Br(),

    # ---------------- Genre Dropdown ----------------
    html.Label("Select Genre"),

    dcc.Dropdown(
        id="genre-dropdown",
        options=[{"label": g, "value": g}
                 for g in sorted(movies["genre"].dropna().unique())],
        value=sorted(movies["genre"].dropna().unique())[0],
        clearable=False
    ),

    html.Br(),

    # ---------------- Country Dropdown ----------------
    html.Label("Select Country"),

    dcc.Dropdown(
        id="country-dropdown",
        options=[{"label": c, "value": c}
                 for c in sorted(movies["country"].dropna().unique())],
        value=sorted(movies["country"].dropna().unique())[0],
        clearable=False
    ),

    html.Br(),

    # ---------------- Graphs ----------------
    dcc.Graph(id="rating-graph"),
    html.Br(),
    dcc.Graph(id="budget-gross-graph")
])

# --------------------------------------------------
# Callback
# --------------------------------------------------
@app.callback(
    Output("rating-graph", "figure"),
    Output("budget-gross-graph", "figure"),
    Input("year-slider", "value"),
    Input("genre-dropdown", "value"),
    Input("country-dropdown", "value")
)
def update_graphs(year_range, selected_genre, selected_country):

    start_year, end_year = year_range

    # Filter data
    filtered = movies[
        (movies["year"] >= start_year) &
        (movies["year"] <= end_year) &
        (movies["genre"] == selected_genre) &
        (movies["country"] == selected_country)
    ]

    # ---------------- Graph 1: Rating ----------------
    avg_score = filtered.groupby("rating")["score"].mean().reset_index()

    fig1 = px.bar(
        avg_score,
        x="rating",
        y="score",
        color="rating",
        text="score",
        title="Average Movie Score by Rating"
    )

    fig1.update_traces(texttemplate="%{text:.2f}", textposition="outside")
    fig1.update_layout(template="plotly_white")

    # ---------------- Graph 2: Budget vs Gross ----------------
    fig2 = px.scatter(
        filtered,
        x="budget",
        y="gross",
        color="rating",
        hover_name="name",
        title="Budget vs Gross Revenue"
    )

    fig2.update_layout(template="plotly_white")

    return fig1, fig2

# --------------------------------------------------
# Run app
# --------------------------------------------------
if __name__ == "__main__":
    app.run(debug=True, port=8051)
```

```python
# Exercise 13
# Improve the appearance of your dashboard by adding at least three of the following:
# Better chart titles
# Different chart colors
# Hover information
# Horizontal bar charts
# Pie charts
# Dashboard styling using html.Div
# Multiple sections or columns
# Appropriate spacing between components
import pandas as pd
import plotly.express as px
from dash import Dash, dcc, html, Input, Output

# --------------------------------------------------
# Load Data
# --------------------------------------------------
movies = pd.read_csv("movies_updated.csv")
movies["year"] = pd.to_numeric(movies["year"], errors="coerce")

# --------------------------------------------------
# App Setup
# --------------------------------------------------
app = Dash(__name__)
app.title = "Improved Movie Dashboard"

# --------------------------------------------------
# Layout (Styled + Sections)
# --------------------------------------------------
app.layout = html.Div(style={"backgroundColor": "#f4f6f9", "padding": "20px"}, children=[

    html.H1(
        "🎬 Movie Analytics Dashboard",
        style={"textAlign": "center", "color": "#2c3e50"}
    ),

    html.Div([
        # LEFT SIDE FILTER PANEL
        html.Div([
            html.H3("Filters"),

            dcc.Dropdown(
                id="genre-dropdown",
                options=[{"label": g, "value": g} for g in sorted(movies["genre"].dropna().unique())],
                value=movies["genre"].dropna().unique()[0],
                clearable=False
            ),

            html.Br(),

            dcc.Dropdown(
                id="country-dropdown",
                options=[{"label": c, "value": c} for c in sorted(movies["country"].dropna().unique())],
                value=movies["country"].dropna().unique()[0],
                clearable=False
            ),

            html.Br(),

            dcc.RangeSlider(
                id="year-slider",
                min=int(movies["year"].min()),
                max=int(movies["year"].max()),
                value=[int(movies["year"].min()), int(movies["year"].max())],
                marks={int(y): str(y) for y in range(int(movies["year"].min()), int(movies["year"].max())+1, 5)},
                step=1
            )

        ], style={"width": "25%", "display": "inline-block", "verticalAlign": "top", "padding": "10px"}),

        # RIGHT SIDE GRAPHS
        html.Div([

            dcc.Graph(id="rating-graph"),

            dcc.Graph(id="budget-gross-graph"),

            dcc.Graph(id="genre-pie-chart")

        ], style={"width": "70%", "display": "inline-block", "padding": "10px"})
    ])
])

# --------------------------------------------------
# Callback
# --------------------------------------------------
@app.callback(
    Output("rating-graph", "figure"),
    Output("budget-gross-graph", "figure"),
    Output("genre-pie-chart", "figure"),
    Input("genre-dropdown", "value"),
    Input("country-dropdown", "value"),
    Input("year-slider", "value")
)
def update_dashboard(genre, country, year_range):

    start_year, end_year = year_range

    filtered = movies[
        (movies["genre"] == genre) &
        (movies["country"] == country) &
        (movies["year"] >= start_year) &
        (movies["year"] <= end_year)
    ]

    # --------------------------------------------------
    # 1. BAR CHART (Better title + color + hover)
    # --------------------------------------------------
    rating_avg = filtered.groupby("rating")["score"].mean().reset_index()

    fig1 = px.bar(
        rating_avg,
        x="rating",
        y="score",
        color="rating",
        text="score",
        hover_data={"score": ":.2f"},
        title="⭐ Average Movie Score by Rating"
    )

    fig1.update_traces(texttemplate="%{text:.2f}", textposition="outside")
    fig1.update_layout(template="plotly_white")

    # --------------------------------------------------
    # 2. SCATTER (Budget vs Gross)
    # --------------------------------------------------
    fig2 = px.scatter(
        filtered,
        x="budget",
        y="gross",
        color="rating",
        hover_name="name",
        title="💰 Budget vs Gross Revenue",
        opacity=0.7
    )

    fig2.update_layout(template="plotly_white")

    # --------------------------------------------------
    # 3. PIE CHART (Genre Distribution)
    # --------------------------------------------------
    genre_count = filtered["genre"].value_counts().reset_index()
    genre_count.columns = ["genre", "count"]

    fig3 = px.pie(
        genre_count,
        names="genre",
        values="count",
        title="🎭 Genre Distribution",
        hole=0.4
    )

    fig3.update_traces(textinfo="percent+label")

    return fig1, fig2, fig3

# --------------------------------------------------
# Run App
# --------------------------------------------------
if __name__ == "__main__":
    app.run(debug=True, port=8051)
```
