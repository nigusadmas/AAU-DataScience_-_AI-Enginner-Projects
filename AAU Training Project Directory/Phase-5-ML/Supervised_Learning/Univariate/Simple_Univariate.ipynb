```python
# =====================================
# Import Required Libraries
# =====================================
import pandas as pd
from sklearn.linear_model import LinearRegression

# =====================================
# Create Dataset
# =====================================
data = {
    "house_size": [100, 120, 150, 180, 200, 220],
    "price": [3000000, 3600000, 4500000, 5500000, 6200000, 7000000]
}

df = pd.DataFrame(data)

# Display Dataset
print(df)

# =====================================
# Select Feature and Target
# =====================================
X = df[["house_size"]]      # Feature
y = df["price"]             # Target

# =====================================
# Create Linear Regression Model
# =====================================
model = LinearRegression()

# =====================================
# Train the Model
# =====================================
model.fit(X, y)

# =====================================
# Predict New House Price
# =====================================
prediction = model.predict([[160]])

print("\nPredicted House Price:")
print(prediction.round(1))

# =====================================
# Model Parameters
# =====================================
print("\nSlope (Coefficient):", model.coef_[0])
print("Intercept:", model.intercept_)

# =====================================
# Accuracy (R² Score)
# =====================================
print("R² Score:", model.score(X, y))
```
