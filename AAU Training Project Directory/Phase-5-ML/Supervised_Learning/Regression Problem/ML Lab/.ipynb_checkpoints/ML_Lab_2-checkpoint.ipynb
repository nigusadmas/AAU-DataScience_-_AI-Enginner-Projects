```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
# Exercise 1: Load and Explore
# 1. Load the dataset using Pandas.
# 2. Display the first five rows.
# 3. Display the dataset shape.
# 4. Display info() .
# 5. Display describe() .
# Questions
# 1. How many observations are in the dataset?
# 2. Which column is the target variable?
# 3. Which column contains text?

# Answers
df = pd.read_csv("auto_mpg_synthetic_dataset.csv")
print(df.head(),"\n")
print(df.shape,"\n")
print(df.info(),"\n")
print(df.describe(),"\n")

#1. There are 300 hundred possible observation in the dataset
#2. The target Variable is "mpg"
#3. The Column that contain text value "car_name" , check it using :-
text_columns = df.select_dtypes(include=['object']).columns
print(text_columns)
```

```python
# Exercise 2: Missing Values
# 1. Check for missing values.
# 2. If any exist, remove them.
# 3. Display the new dataset shape.
# Questions
# 1. Which columns contain missing values?
# 2. Why should missing values be handled before training?

# Answer
print(df.isnull().sum())
df=df.dropna()
df

#1. There is no column that contain missing value
#2. Machine Learning algorithms cannot learn from empty values.
    # Missing data can reduce model accuracy.
    # It may cause errors during training.
    # Handled missing value produces more reliable predictions.
```

```python
# Exercise 3: Exploratory Data Analysis
# Create:
#     Histogram of MPG
#     Box plot of MPG
#     Correlation matrix
#     Correlation bar chart
#     Scatter plots:
#         MPG vs Weight
#         MPG vs Horsepower
#         MPG vs Cylinders
#         MPG vs Displacement
#         MPG vs Acceleration
#         MPG vs Model Year
# For each scatter plot answer:
#     Is the relationship positive or negative?
#     Is it strong or weak?
#     Are there any outliers?

# 1
plt.figure(figsize=(8,5))
plt.hist(df["mpg"], bins=20, edgecolor="black")
plt.title("Distribution of MPG")
plt.xlabel("MPG")
plt.ylabel("Frequency")
plt.show()

# 2
plt.figure(figsize=(6,4))
sns.boxplot(df["mpg"])
plt.title("Box Plot of MPG")
plt.show()

# 3
numeric_df = df.drop(columns=["car_name"])
corr = numeric_df.corr()
plt.figure(figsize=(8,6))
sns.heatmap(corr,
            annot=True,
            cmap="coolwarm",
            fmt=".2f")
plt.title("Correlation Matrix")
plt.show()

# 4
print("MPG vs Weight")
plt.figure(figsize=(6,4))
sns.scatterplot(x="weight",
                y="mpg",
                data=df)
plt.title("MPG vs Weight")
plt.show()


print("MPG vs Horsepower")
plt.figure(figsize=(6,4))
sns.scatterplot(x="horsepower",
                y="mpg",
                data=df)
plt.title("MPG vs Horsepower")
plt.show()


print("MPG vs Cylinders")
plt.figure(figsize=(6,4))
sns.scatterplot(x="cylinders",
                y="mpg",
                data=df)
plt.title("MPG vs Cylinders")
plt.show()


print("MPG vs Displacement")
plt.figure(figsize=(6,4))
sns.scatterplot(x="displacement",
                y="mpg",
                data=df)
plt.title("MPG vs Displacement")
plt.show()


print("MPG vs Acceleration")
plt.figure(figsize=(6,4))
sns.scatterplot(x="acceleration",
                y="mpg",
                data=df)
plt.title("MPG vs Acceleration")
plt.show()


print("MPG vs Model Year")
plt.figure(figsize=(6,4))
sns.scatterplot(x="model_year",
                y="mpg",
                data=df)

plt.title("MPG vs Model Year")
plt.show()
from sklearn.linear_model import LinearRegression
import numpy as np
def interpret_correlation(r):
    if r > 0.8:
        return "Very strong positive"
    elif r > 0.6:
        return "Strong positive"
    elif r > 0.4:
        return "Moderate positive"
    elif r > 0:
        return "Weak positive"
    elif r > -0.4:
        return "Weak negative"
    elif r > -0.6:
        return "Moderate negative"
    elif r > -0.8:
        return "Strong negative"
    else:
        return "Very strong negative"
features = [
    "weight",
    "horsepower",
    "cylinders",
    "displacement",
    "acceleration",
    "model_year"
]
for feature in features:
    # -------------------------
    # 1. Correlation
    # -------------------------
    r = df["mpg"].corr(df[feature])
    print("\nFeature:", feature)
    print("Correlation:", round(r,3))
    print("Relationship:", interpret_correlation(r))
    print("----------------------")
```
