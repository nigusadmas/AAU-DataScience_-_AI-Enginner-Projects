```python
# =====================================
# Import Required Libraries
# =====================================
import pandas as pd
from sklearn.linear_model import LinearRegression

# =====================================
# Create Dataset
# =====================================
data = {
    "house_size": [100,120,150,180,200,220],
    "bedrooms": [2,3,3,4,4,5],
    "age": [20,15,10,8,5,2],
    "location_score": [5,6,7,8,9,10],
    "price": [3000000,3600000,4500000,5500000,6200000,7000000]
}

df = pd.DataFrame(data)

# Display Dataset
# print(df)

# =====================================
# Select Multiple Features
# =====================================
X = df[[
    "house_size",
    "bedrooms",
    "age",
    "location_score"
]]

# =====================================
# Target Variable
# =====================================
y = df["price"]

# =====================================
# Create Model
# =====================================
model = LinearRegression()

# =====================================
# Train Model
# =====================================
model.fit(X, y)

# =====================================
# Predict New House
# =====================================
prediction = model.predict([[160,4,5,8]])

print("\nPredicted House Price:")
print(prediction)

# =====================================
# Coefficients
# =====================================
print("\nCoefficients:")
for feature, coef in zip(X.columns, model.coef_):
    print(feature, ":", coef)

# =====================================
# Intercept
# =====================================
print("\nIntercept:", model.intercept_)

# =====================================
# Accuracy
# =====================================
print("R² Score:", model.score(X, y))
```
