```python
# Dataset 2: Orders
orders = pd.DataFrame({
"Order_ID": [101,102,103,104,105,106,107],
"Customer_ID": ["C001","C002","C001","C004","C007","C006","C006"],"Product":
["Laptop","Phone","Mouse","Keyboard","Tablet","Monitor","Printer"],"Amount": [1200,800,45,150,650,700,300]
})
orders
```

```python
# Exercise 2 — Order Investigation
# The finance department wants to review every order, even if customer information is
# unavailable.
# Create a report showing:
# Order ID
# Customer ID
# Customer Name
# Product
# Amount
# Questions
# 1. Which order has missing customer information?
# order_Id=105
# 2. Why is the customer information missing?
# Orders table has C007
# But customers table does NOT contain C007
# So join cannot find matching row
# 3. How many registered customers do not appear in this report?
# Answer: 0 customers
report2 = pd.merge(customers, orders, on="Customer_ID", how="right")
report2 = report2[["Order_ID","Customer_ID","Customer_Name","Product","Amount"]]
report2
```
