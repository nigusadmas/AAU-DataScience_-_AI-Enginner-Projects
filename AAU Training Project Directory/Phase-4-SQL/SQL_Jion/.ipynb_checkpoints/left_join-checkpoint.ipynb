```python
# Dataset 1: Customers
import pandas as pd
customers = pd.DataFrame({
"Customer_ID": ["C001", "C002", "C003", "C004", "C005", "C006"],
"Customer_Name": ["Abel", "Sara", "John", "Helen", "Samuel","David"],
"City": ["Addis Ababa", "Adama", "Jimma", "Bahir Dar", "Hawassa","Dire Dawa"]
})
customers
```

```python
# Dataset 2: Orders
orders = pd.DataFrame({
"Order_ID": [101,102,103,104,105,106,107],
"Customer_ID": ["C001","C002","C001","C004","C007","C006","C006"],"Product":
["Laptop","Phone","Mouse","Keyboard","Tablet","Monitor","Printer"],"Amount": [1200,800,45,150,650,700,300]
})
orders
```

```python
# Exercise 3 — Customer Engagement
# Report
# The marketing team wants a list of all registered customers, including those who have
# never purchased anything.
# Display:
# Customer Name,City,Product,Amount
# Questions
# 1. Which customer has never placed an order?
# 2. What values appear for that customer's Product and Amount?
# 3. Why do those values appear?


report3 = pd.merge(customers, orders, on="Customer_ID", how="left")
report3 = report3[["Customer_Name","City","Product","Amount"]]
report3
```
