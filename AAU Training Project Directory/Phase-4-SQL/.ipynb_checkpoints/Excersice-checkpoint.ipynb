```python
import pandas as pd
```

```python
students = pd.DataFrame({
"Student_ID": [1, 2, 3, 4],
"Student_Name": ["Abel", "Sara", "John", "Helen"]
})
students
```

```python
enrollments = pd.DataFrame({
"Student_ID": [1, 2, 5],
"Course": ["Python", "SQL", "Machine Learning"]
})
enrollments
```

```python
# inner join
inner=pd.merge(
    students,
    enrollments,
    on="Student_ID",
    how="inner"
)
inner
```

```python
# left join
left=pd.merge(
    students,
    enrollments,
    on="Student_ID",
    how="left"
)
left
```

```python
# right join
right=pd.merge(
    students,
    enrollments,
    on="Student_ID",
    how="right"
)
right
```

```python
# outer join
outer=pd.merge(
    students,
    enrollments,
    on="Student_ID",
    how="outer"
)
outer
```
