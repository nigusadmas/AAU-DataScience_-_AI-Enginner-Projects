```python
# iteration in pandas dataframe 
import pandas as pd
data = {
    "Name": ["Abebe", "Hana", "Kebede"],
    "Age": [20, 22, 25],
    "Salary": [12000, 18000, 15000]
}
df = pd.DataFrame(data)
df

for col_name , col_data in df.items():
    print(col_name)
    print("=============")
    print(col_data)
    print("\n")
print("$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$")
for column, value in df.items():
    print("Column Name :", column)
    print(value)
    print("----------------")
```

```python
for i, r in df.iterrows():
    print(i)
    print(r)
```
