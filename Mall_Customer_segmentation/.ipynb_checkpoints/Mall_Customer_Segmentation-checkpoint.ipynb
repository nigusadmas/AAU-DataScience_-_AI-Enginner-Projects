```python
# ============================================================
# K-MEANS CLUSTERING: MALL CUSTOMER SEGMENTATION
# ============================================================

# Import required libraries
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import (
    silhouette_score,
    calinski_harabasz_score,
    davies_bouldin_score
)

# Ignore warnings
warnings.filterwarnings("ignore")

# Set visualization style
plt.style.use("seaborn-v0_8-whitegrid")
sns.set_palette("husl")

print("=" * 80)
print("K-MEANS CLUSTERING: MALL CUSTOMER SEGMENTATION")
print("=" * 80)

# ============================================================
# STEP 1: LOAD AND EXPLORE THE DATASET
# ============================================================

print("\nSTEP 1: Loading and Exploring the Dataset")
print("-" * 50)
# Online backup dataset
online_path = "https://raw.githubusercontent.com/plotly/datasets/master/mall_customers_2019.csv"

try:
    # Try loading local file
    df = pd.read_csv("dataset/Mall_Customers.csv")
    print("✅ Dataset loaded successfully from local file!")

except FileNotFoundError:
    # If local file is missing, load from GitHub
    df = pd.read_csv(online_path)
    print("✅ Dataset loaded successfully from online source!")

# Display first five rows
print("\nFirst 5 Rows:")
print(df.head(10))

# Dataset information
print("\nDataset Information:")
print(df.info())

# Statistical summary
print("\nStatistical Summary:")
print(df.describe())

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Dataset shape
print(f"\nDataset Shape: {df.shape}")
```

```python
# ============================================================
# STEP 2: EXPLORATORY DATA ANALYSIS (EDA)
# ============================================================

print("\n\nSTEP 2: Exploratory Data Analysis")
print("-" * 50)

# Create figure for EDA
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# ------------------------------------------------------------
# Distribution of Age
# ------------------------------------------------------------
axes[0, 0].hist(df["Age"], bins=20, edgecolor="black", alpha=0.7)
axes[0, 0].set_xlabel("Age")
axes[0, 0].set_ylabel("Frequency")
axes[0, 0].set_title(
    "Distribution of Customer Ages",
    fontsize=12,
    fontweight="bold"
)
axes[0, 0].axvline(
    df["Age"].mean(),
    color="red",
    linestyle="dashed",
    linewidth=2,
    label=f"Mean: {df['Age'].mean():.1f}"
)
axes[0, 0].legend()

# ------------------------------------------------------------
# Distribution of Annual Income
# ------------------------------------------------------------
axes[0, 1].hist(
    df["Annual Income (k$)"],
    bins=20,
    edgecolor="black",
    alpha=0.7,
    color="green"
)
axes[0, 1].set_xlabel("Annual Income (k$)")
axes[0, 1].set_ylabel("Frequency")
axes[0, 1].set_title(
    "Distribution of Annual Income",
    fontsize=12,
    fontweight="bold"
)
axes[0, 1].axvline(
    df["Annual Income (k$)"].mean(),
    color="red",
    linestyle="dashed",
    linewidth=2,
    label=f"Mean: ${df['Annual Income (k$)'].mean():.1f}k"
)
axes[0, 1].legend()

# ------------------------------------------------------------
# Distribution of Spending Score
# ------------------------------------------------------------
axes[1, 0].hist(
    df["Spending Score (1-100)"],
    bins=20,
    edgecolor="black",
    alpha=0.7,
    color="orange"
)
axes[1, 0].set_xlabel("Spending Score (1-100)")
axes[1, 0].set_ylabel("Frequency")
axes[1, 0].set_title(
    "Distribution of Spending Scores",
    fontsize=12,
    fontweight="bold"
)
axes[1, 0].axvline(
    df["Spending Score (1-100)"].mean(),
    color="red",
    linestyle="dashed",
    linewidth=2,
    label=f"Mean: {df['Spending Score (1-100)'].mean():.1f}"
)
axes[1, 0].legend()

# ------------------------------------------------------------
# Gender Distribution (if available)
# ------------------------------------------------------------
if "Gender" in df.columns:
    gender_counts = df["Gender"].value_counts()

    axes[1, 1].pie(
        gender_counts.values,
        labels=gender_counts.index,
        autopct="%1.1f%%",
        startangle=90
    )

    axes[1, 1].set_title(
        "Gender Distribution",
        fontsize=12,
        fontweight="bold"
    )
else:
    axes[1, 1].scatter(
        df["Annual Income (k$)"],
        df["Spending Score (1-100)"],
        alpha=0.6
    )

    axes[1, 1].set_xlabel("Annual Income (k$)")
    axes[1, 1].set_ylabel("Spending Score (1-100)")
    axes[1, 1].set_title(
        "Income vs Spending Score",
        fontsize=12,
        fontweight="bold"
    )

plt.tight_layout()
plt.savefig("figures/eda_visualizations.png", dpi=300, bbox_inches="tight")
plt.show()

# ============================================================
# Relationship Between Features
# ============================================================

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# ------------------------------------------------------------
# Income vs Spending Score
# ------------------------------------------------------------
scatter = axes[0].scatter(
    df["Annual Income (k$)"],
    df["Spending Score (1-100)"],
    c=df["Age"],
    cmap="viridis",
    alpha=0.6,
    s=50
)

axes[0].set_xlabel("Annual Income (k$)", fontsize=11)
axes[0].set_ylabel("Spending Score (1-100)", fontsize=11)
axes[0].set_title(
    "Income vs Spending Score (Colored by Age)",
    fontsize=12,
    fontweight="bold"
)

plt.colorbar(scatter, ax=axes[0], label="Age")

# ------------------------------------------------------------
# Age vs Spending Score
# ------------------------------------------------------------
axes[1].scatter(
    df["Age"],
    df["Spending Score (1-100)"],
    alpha=0.6,
    color="steelblue",
    s=50
)

axes[1].set_xlabel("Age", fontsize=11)
axes[1].set_ylabel("Spending Score (1-100)", fontsize=11)
axes[1].set_title(
    "Age vs Spending Score",
    fontsize=12,
    fontweight="bold"
)

plt.tight_layout()
plt.savefig("figures/feature_relationships.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# ============================================================
# STEP 3: DATA PREPROCESSING
# ============================================================

print("\n\nSTEP 3: Data Preprocessing")
print("-" * 50)

# ------------------------------------------------------------
# Select features for clustering
# Using Annual Income and Spending Score because they are the
# most informative features for customer segmentation.
# ------------------------------------------------------------
features = ["Annual Income (k$)", "Spending Score (1-100)"]

# Create feature matrix
X = df[features].values

print(f"Selected Features: {features}")
print(f"Feature Matrix Shape: {X.shape}")

# ------------------------------------------------------------
# Standardize the features
# K-Means is distance-based, so feature scaling is essential.
# ------------------------------------------------------------
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("✓ Features standardized successfully (mean = 0, std = 1)")

# ------------------------------------------------------------
# Display the first 10 rows of the scaled data
# ------------------------------------------------------------
print("\nSample of Scaled Features (First 10 Rows):")

scaled_df = pd.DataFrame(
    X_scaled[:10],
    columns=["Scaled Income", "Scaled Spending Score"]
)

print(scaled_df)
```

```python
# ============================================================
# STEP 4: FINDING THE OPTIMAL NUMBER OF CLUSTERS
# (Elbow Method & Silhouette Score)
# ============================================================

print("\n\nSTEP 4: Finding the Optimal Number of Clusters")
print("-" * 50)

# ------------------------------------------------------------
# Calculate WCSS (Within-Cluster Sum of Squares)
# and Silhouette Score for different values of k
# ------------------------------------------------------------
wcss = []
silhouette_scores = []

k_range = range(2, 11)

for k in k_range:
    kmeans = KMeans(
        n_clusters=k,
        init="k-means++",
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_scaled)

    # Store WCSS (Inertia)
    wcss.append(kmeans.inertia_)

    # Store Silhouette Score
    silhouette_scores.append(
        silhouette_score(X_scaled, kmeans.labels_)
    )

# ============================================================
# Create Visualization
# ============================================================

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# ------------------------------------------------------------
# Elbow Method Plot
# ------------------------------------------------------------
axes[0].plot(
    k_range,
    wcss,
    "bo-",
    linewidth=2,
    markersize=8
)

axes[0].axvline(
    x=5,
    color="red",
    linestyle="--",
    linewidth=2,
    label="Optimal k = 5"
)

axes[0].set_xlabel("Number of Clusters (k)", fontsize=12)
axes[0].set_ylabel("Within-Cluster Sum of Squares (WCSS)", fontsize=12)
axes[0].set_title(
    "Elbow Method for Optimal k",
    fontsize=14,
    fontweight="bold"
)
axes[0].grid(True, alpha=0.3)
axes[0].legend()

# ------------------------------------------------------------
# Silhouette Score Plot
# ------------------------------------------------------------
axes[1].plot(
    k_range,
    silhouette_scores,
    "go-",
    linewidth=2,
    markersize=8
)

axes[1].axvline(
    x=5,
    color="red",
    linestyle="--",
    linewidth=2,
    label="Optimal k = 5"
)

axes[1].set_xlabel("Number of Clusters (k)", fontsize=12)
axes[1].set_ylabel("Silhouette Score", fontsize=12)
axes[1].set_title(
    "Silhouette Score for Different k",
    fontsize=14,
    fontweight="bold"
)
axes[1].grid(True, alpha=0.3)
axes[1].legend()

plt.tight_layout()
plt.savefig(
    "figures/optimal_k_determination.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()

# ============================================================
# Select the Optimal Number of Clusters
# ============================================================

optimal_k = 5

print(f"\nOptimal number of clusters determined: k = {optimal_k}")
print("Reason: Based on the elbow point and a high silhouette score.")
```

```python
# ============================================================
# STEP 5: APPLYING K-MEANS CLUSTERING
# ============================================================

print("\n\nSTEP 5: Applying K-Means Clustering")
print("-" * 50)

# ------------------------------------------------------------
# Apply K-Means using the optimal number of clusters
# ------------------------------------------------------------
kmeans = KMeans(
    n_clusters=optimal_k,
    init="k-means++",
    random_state=42,
    n_init=10
)

# Fit the model and predict cluster labels
y_kmeans = kmeans.fit_predict(X_scaled)

# ------------------------------------------------------------
# Add cluster labels to the original dataset
# ------------------------------------------------------------
df["Cluster"] = y_kmeans

# ------------------------------------------------------------
# Convert cluster centers back to the original scale
# ------------------------------------------------------------
centers_original = scaler.inverse_transform(kmeans.cluster_centers_)

print("✓ K-Means clustering completed successfully!")

# ------------------------------------------------------------
# Display cluster distribution
# ------------------------------------------------------------
print("\nCluster Distribution:")

cluster_counts = df["Cluster"].value_counts().sort_index()

for i in range(optimal_k):
    print(
        f"Cluster {i}: "
        f"{cluster_counts[i]} customers "
        f"({cluster_counts[i] / len(df) * 100:.1f}%)"
    )

# ------------------------------------------------------------
# Display cluster centers
# ------------------------------------------------------------
cluster_centers = pd.DataFrame(
    centers_original,
    columns=["Annual Income (k$)", "Spending Score (1-100)"]
)

cluster_centers.index = [
    f"Cluster {i}" for i in range(optimal_k)
]

print("\nCluster Centers (Original Scale):")
print(cluster_centers.round(2))
```

```python
# ============================================================
# STEP 6: CLUSTER VISUALIZATION
# ============================================================

print("\n\nSTEP 6: Cluster Visualization")
print("-" * 50)

# ------------------------------------------------------------
# Define colors for the clusters
# ------------------------------------------------------------
colors = [
    "#FF6B6B",
    "#4ECDC4",
    "#45B7D1",
    "#96CEB4",
    "#FFEAA7"
]

# ============================================================
# Create visualization
# ============================================================

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# ------------------------------------------------------------
# Plot clusters using original feature values
# ------------------------------------------------------------
for i in range(optimal_k):

    cluster_data = df[df["Cluster"] == i]

    axes[0].scatter(
        cluster_data["Annual Income (k$)"],
        cluster_data["Spending Score (1-100)"],
        color=colors[i],
        label=f"Cluster {i}",
        alpha=0.7,
        s=60,
        edgecolors="black",
        linewidth=0.5
    )

# Plot cluster centroids
axes[0].scatter(
    centers_original[:, 0],
    centers_original[:, 1],
    color="red",
    marker="X",
    s=300,
    edgecolors="black",
    linewidth=2,
    label="Centroids"
)

axes[0].set_xlabel("Annual Income (k$)", fontsize=12)
axes[0].set_ylabel("Spending Score (1-100)", fontsize=12)
axes[0].set_title(
    f"Customer Segments (k = {optimal_k})",
    fontsize=14,
    fontweight="bold"
)
axes[0].legend(loc="best")
axes[0].grid(True, alpha=0.3)

# ------------------------------------------------------------
# Plot clusters using standardized data
# ------------------------------------------------------------
for i in range(optimal_k):

    cluster_data_scaled = X_scaled[y_kmeans == i]

    axes[1].scatter(
        cluster_data_scaled[:, 0],
        cluster_data_scaled[:, 1],
        color=colors[i],
        label=f"Cluster {i}",
        alpha=0.7,
        s=60,
        edgecolors="black",
        linewidth=0.5
    )

# Plot scaled centroids
axes[1].scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    color="red",
    marker="X",
    s=300,
    edgecolors="black",
    linewidth=2,
    label="Centroids"
)

axes[1].set_xlabel("Scaled Annual Income", fontsize=12)
axes[1].set_ylabel("Scaled Spending Score", fontsize=12)
axes[1].set_title(
    "Customer Segments (Standardized Data)",
    fontsize=14,
    fontweight="bold"
)
axes[1].legend(loc="best")
axes[1].grid(True, alpha=0.3)

# ============================================================
# Save and display the figure
# ============================================================

plt.tight_layout()
plt.savefig(
    "figures/cluster_visualization.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

```python
# ============================================================
# STEP 7: CLUSTER QUALITY EVALUATION
# ============================================================

print("\n\nSTEP 7: Cluster Quality Evaluation")
print("-" * 50)

# ------------------------------------------------------------
# Calculate evaluation metrics
# ------------------------------------------------------------
sil_score = silhouette_score(X_scaled, y_kmeans)
calinski_score = calinski_harabasz_score(X_scaled, y_kmeans)
davies_bouldin = davies_bouldin_score(X_scaled, y_kmeans)

print("CLUSTERING PERFORMANCE METRICS")
print("=" * 50)

print(f"Silhouette Score: {sil_score:.4f}")
print("   Range: -1 to 1 (Higher is Better)")
print("   Values above 0.50 generally indicate good clustering.\n")

print(f"Calinski-Harabasz Index: {calinski_score:.2f}")
print("   Higher values indicate better-defined clusters.\n")

print(f"Davies-Bouldin Index: {davies_bouldin:.4f}")
print("   Lower values indicate better cluster separation (0 is ideal).")

# ------------------------------------------------------------
# Interpretation
# ------------------------------------------------------------
print("\nQUALITY INTERPRETATION")
print("-" * 35)

if sil_score > 0.5:
    print("Excellent silhouette score - clusters are well separated.")
elif sil_score > 0.3:
    print("Good silhouette score - reasonable cluster structure.")
else:
    print("Low silhouette score - clusters may overlap significantly.")

# ------------------------------------------------------------
# Silhouette Analysis Visualization
# ------------------------------------------------------------
from sklearn.metrics import silhouette_samples

fig, ax = plt.subplots(1, 2, figsize=(14, 6))

# Compute silhouette values for each sample
silhouette_vals = silhouette_samples(X_scaled, y_kmeans)

y_lower = 10

for i in range(optimal_k):

    ith_cluster_silhouette = silhouette_vals[y_kmeans == i]
    ith_cluster_silhouette.sort()

    size_cluster = ith_cluster_silhouette.shape[0]
    y_upper = y_lower + size_cluster

    color = colors[i]

    ax[0].fill_betweenx(
        np.arange(y_lower, y_upper),
        0,
        ith_cluster_silhouette,
        facecolor=color,
        edgecolor="black",
        alpha=0.7,
        linewidth=0.5
    )

    ax[0].text(
        -0.05,
        y_lower + size_cluster / 2,
        str(i),
        fontsize=12
    )

    y_lower = y_upper + 10

# Average silhouette score
ax[0].axvline(
    x=sil_score,
    color="red",
    linestyle="--",
    linewidth=2,
    label=f"Average = {sil_score:.3f}"
)

ax[0].set_xlabel("Silhouette Coefficient", fontsize=12)
ax[0].set_ylabel("Clusters", fontsize=12)
ax[0].set_title(
    "Silhouette Plot for Clusters",
    fontsize=14,
    fontweight="bold"
)
ax[0].set_xlim([-0.1, 1])
ax[0].legend()
ax[0].grid(True, alpha=0.3)

# ------------------------------------------------------------
# Cluster Size Comparison
# ------------------------------------------------------------
cluster_sizes = df["Cluster"].value_counts().sort_index()

bars = ax[1].bar(
    range(optimal_k),
    cluster_sizes.values,
    color=colors,
    edgecolor="black"
)

ax[1].set_xlabel("Cluster", fontsize=12)
ax[1].set_ylabel("Number of Customers", fontsize=12)
ax[1].set_title(
    "Cluster Sizes",
    fontsize=14,
    fontweight="bold"
)
ax[1].set_xticks(range(optimal_k))
ax[1].grid(True, alpha=0.3, axis="y")

# Add labels above bars
for bar, size in zip(bars, cluster_sizes.values):
    ax[1].text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 2,
        str(size),
        ha="center",
        va="bottom",
        fontsize=10,
        fontweight="bold"
    )

# ------------------------------------------------------------
# Save and display the figure
# ------------------------------------------------------------
plt.tight_layout()
plt.savefig(
    "figures/cluster_quality_evaluation.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

```python
# ============================================================
# STEP 8: CLUSTER PROFILING AND BUSINESS INSIGHTS
# ============================================================

print("\n\nSTEP 8: Cluster Profiling and Business Insights")
print("-" * 50)

# ------------------------------------------------------------
# Calculate Cluster Profiles
# ------------------------------------------------------------
cluster_profiles = df.groupby("Cluster")[
    ["Age", "Annual Income (k$)", "Spending Score (1-100)"]
].agg(["mean", "std", "count"])

print("\nCLUSTER PROFILES")
print("=" * 80)
print(cluster_profiles)

# ------------------------------------------------------------
# Display summary for each cluster
# ------------------------------------------------------------
for i in range(optimal_k):

    cluster_data = df[df["Cluster"] == i]

    print(
        f"\nCluster {i}: "
        f"{len(cluster_data)} customers "
        f"({len(cluster_data) / len(df) * 100:.1f}%)"
    )

    print(f" • Average Age: {cluster_data['Age'].mean():.1f} years")
    print(
        f" • Average Income: "
        f"${cluster_data['Annual Income (k$)'].mean():.1f}k"
    )
    print(
        f" • Average Spending Score: "
        f"{cluster_data['Spending Score (1-100)'].mean():.1f}/100"
    )

# ============================================================
# BUSINESS INSIGHTS
# ============================================================

print("\n\nBUSINESS INSIGHTS & CUSTOMER SEGMENTS")
print("=" * 80)

# Cluster centers in original scale
cluster_centers_df = pd.DataFrame(
    centers_original,
    columns=["Income", "Spending"]
)

cluster_centers_df["Cluster"] = range(optimal_k)

# ------------------------------------------------------------
# Interpret each cluster
# ------------------------------------------------------------
for _, row in cluster_centers_df.iterrows():

    cluster = int(row["Cluster"])
    income = row["Income"]
    spending = row["Spending"]

    if income > 70 and spending > 60:

        print(
            f"\nCluster {cluster}: HIGH VALUE CUSTOMERS"
        )
        print(
            f"Income: ${income:.0f}k | "
            f"Spending Score: {spending:.0f}"
        )
        print(
            "Strategy: Premium loyalty programs, "
            "VIP treatment, and exclusive offers."
        )

    elif income > 70 and spending <= 60:

        print(
            f"\nCluster {cluster}: TARGETABLE WEALTHY CUSTOMERS"
        )
        print(
            f"Income: ${income:.0f}k | "
            f"Spending Score: {spending:.0f}"
        )
        print(
            "Strategy: Encourage spending through premium "
            "products and personalized promotions."
        )

    elif income <= 70 and spending > 60:

        print(
            f"\nCluster {cluster}: ASPIRATIONAL SHOPPERS"
        )
        print(
            f"Income: ${income:.0f}k | "
            f"Spending Score: {spending:.0f}"
        )
        print(
            "Strategy: Offer installment plans, discounts, "
            "and bundled products."
        )

    elif income <= 40 and spending <= 40:

        print(
            f"\nCluster {cluster}: FRUGAL CUSTOMERS"
        )
        print(
            f"Income: ${income:.0f}k | "
            f"Spending Score: {spending:.0f}"
        )
        print(
            "Strategy: Promote affordable products, "
            "budget-friendly deals, and essential items."
        )

    else:

        print(
            f"\nCluster {cluster}: MODERATE CUSTOMERS"
        )
        print(
            f"Income: ${income:.0f}k | "
            f"Spending Score: {spending:.0f}"
        )
        print(
            "Strategy: Increase engagement through "
            "cross-selling and up-selling campaigns."
        )

# ============================================================
# END OF ANALYSIS
# ============================================================

print("\n" + "=" * 80)
print("K-MEANS CLUSTERING ANALYSIS COMPLETED SUCCESSFULLY!")
print("=" * 80)
```
