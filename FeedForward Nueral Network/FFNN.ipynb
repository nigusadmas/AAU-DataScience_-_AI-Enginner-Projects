```python
# ReLU Activation
# negative number → 0
# positive number → unchanged
# for example:-
    # -6 ->0
    #  6 ->6
```

```python
# create a python function relu() that return:-
    # 0 if x is negative
    #  x if x is positive

def relu(x):
    if x < 0:
        return 0
    else:
        return x
print(relu(-5))  # Output: 0
print(relu(3))   # Output: 3
print(relu(10))  # Output: 10
print(relu(-10))  # Output: 0
```

```python
# Another option
def relu(x):
    return max(0, x)
print(relu(-5))
print(relu(-2))
print(relu(0))
print(relu(3))
print(relu(10))
```

```python
# Part 3: Build a Complete Neuron
def neuron(x1, x2, w1, w2, bias):
    # Step 1: weighted sum
    z = x1 * w1 + x2 * w2 + bias
    
    # Step 2: activation
    output = relu(z)
    
    # Step 3: return
    return output
    
output = neuron(2, 3,0.5, 0.2,1)
print(output)
```

```python
# Part 4: Create Two Neurons
x1 = 2
x2 = 3
# Neuron 1
h1 = neuron(x1, x2,0.5, 0.2,1)

# Neuron 2
h2 = neuron(x1, x2,-0.3, 0.8,0.5)
print("Neuron 1:", h1)
print("Neuron 2:", h2)
```

```python
# Part 5: Create a Hidden Layer
# Print the hidden-layer outputs.
print("Hidden neuron 1:", h1)
print("Hidden neuron 2:", h2)
# Then answer:
 # Why do we call these neurons "hidden" neurons?
    # Solution
        # They are called hidden neurons because they are between the input layer and 
        # the output layer.
# Input Layer
# ↓
# Hidden Layer
# ↓
# Output Layer
```

```python
# Part 6: Create the Output Neuron
output = neuron(h1,h2,0.4, 0.6,0.2)
print("Final output:", output)
```

```python
# Part 8: Build the Complete Network
# =========================
# Activation Function
# =========================
def relu(x):
    return max(0, x)
    
# =========================
# Neuron
# =========================
def neuron(x1, x2, w1, w2, bias):
    z = x1 * w1 + x2 * w2 + bias
    return relu(z)
    
# =========================
# Input
# =========================
x1 = 2
x2 = 3

# =========================
# Hidden Layer
# =========================
h1 = neuron(x1, x2,0.5, 0.2,1)
h2 = neuron(x1, x2,-0.3, 0.8,0.5)

# =========================
# Output Layer
# =========================
output = neuron(h1, h2,0.4, 0.6,0.2)

# =========================
# Results
# =========================
print("Input 1:", x1)
print("Input 2:", x2)
print("Hidden neuron 1:", h1)
print("Hidden neuron 2:", h2)
print("Final output:", output)
```

```python
# Part 9: Add third Hidden Neuron

h3 = neuron(x1,x2,0.7,-0.4,0.3)
print("h1:", h1)
print("h2:", h2)
print("h3:", h3)
```

```python
# Part 10: Build a Network With 3 Hidden Neurons
def output_neuron(h1, h2, h3, w1, w2, w3, bias):
    z = (h1 * w1+ h2 * w2+ h3 * w3+ bias)
    output = relu(z)
    return output
    
prediction = output_neuron(h1,h2,h3,0.4,0.6,0.2,0.1)
print("Prediction:", prediction)
```

```python
# Part 11: Complete 3-Neuron Network
import numpy as np

# =========================
# Activation Function (ReLU)
# =========================
def relu(x):
    return max(0, x)


# =========================
# Two-input Neuron
# =========================
def neuron(x1, x2, w1, w2, bias):
    z = x1 * w1 + x2 * w2 + bias
    return relu(z)


# =========================
# Output Neuron
# =========================
def output_neuron(h1, h2, h3, w1, w2, w3, bias):
    z = (
        h1 * w1
        + h2 * w2
        + h3 * w3
        + bias
    )
    return relu(z)


# =========================
# Input
# =========================
x1 = 2
x2 = 3


# =========================
# Hidden Layer
# =========================
h1 = neuron(
    x1, x2,
    0.5, 0.2,
    1
)

h2 = neuron(
    x1, x2,
    -0.3, 0.8,
    0.5
)

h3 = neuron(
    x1, x2,
    0.7, -0.4,
    0.3
)


# =========================
# Output Layer
# =========================
prediction = output_neuron(
    h1,
    h2,
    h3,
    0.4,
    0.6,
    0.2,
    0.1
)


# =========================
# Print Results
# =========================
print("Input 1:", x1)
print("Input 2:", x2)

print("\nHidden Layer:")
print("h1:", h1)
print("h2:", h2)
print("h3:", h3)

print("\nOutput Layer:")
print("Prediction:", prediction)
```
