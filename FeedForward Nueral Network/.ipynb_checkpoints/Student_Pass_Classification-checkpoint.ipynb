# Exercise 1 — Predict Whether a Student Will Pass

## Goal

Build a **binary classification neural network** that predicts whether a student will pass based on:

- `hours_studied`
- `attendance`
- `previous_score`

Target:

- `0` = Fail
- `1` = Pass

### Neural network

```text
hours_studied ───┐
attendance ──────┼──> Dense → ReLU → Dense → Sigmoid → Pass/Fail
previous_score ──┘
```
